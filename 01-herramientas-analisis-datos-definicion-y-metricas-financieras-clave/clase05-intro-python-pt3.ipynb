## 3. Manejo de excepciones (10 min)

* ¿Qué es una excepción?
* Uso de `try`, `except`, `else`, `finally`.

```python
# División segura con manejo de errores
def division_segura(a, b):
    try:
        resultado = a / b
        print(f"Resultado: {resultado}")
    except ZeroDivisionError:
        print("No se puede dividir entre cero.")
    except Exception as e:
        print("Error inesperado:", e)
    else:
        print("División exitosa.")
    finally:
        print("Fin del intento.")

division_segura(10, 0)
```

## 4. Ciclos o bucles (20 min)

* `for` vs `while`: diferencias, cuándo usar cada uno.
* La función `range()`.
* Iteración sobre listas, strings y rangos.

```python
# for loop con range
for i in range(1, 11):
    print(f"3 x {i} = {3*i}")

# while loop
n = 0
while n < 5:
    print("Repetición número:", n)
    n += 1

# Iterar sobre lista
grupos = ["A", "B", "C"]
for grupo in grupos:
    print("Grupo:", grupo)
```

## 5. Funciones: definición, argumentos y retorno (20 min)

* ¿Qué es una función? Ventajas para la reutilización y organización del código.
* Sintaxis básica: `def`, argumentos, valor de retorno.
* Funciones con argumentos por defecto, argumentos variables (`*args`, `**kwargs`).
* Diferencia entre variables locales y globales.

```python
# Definición básica de función
def saludo(nombre):
    return f"Hola, {nombre}!"

print(saludo("Ana"))

# Factorial (recursivo)
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))

# Contar palabras en un string usando diccionario
def contar_palabras(texto):
    palabras = texto.lower().split()
    conteo = {}
    for palabra in palabras:
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo

texto = "el dato es el nuevo oro y el dato es poder"
print(contar_palabras(texto))
```

## 6. Introducción a Clases y Objetos (15 min)

* ¿Qué es una clase? ¿Qué es un objeto? Analogía con cosas del mundo real (ejemplo: auto).
* Sintaxis básica: `class`, atributos, métodos, constructor `__init__`.
* Instanciación y uso de objetos.
* Breve mención del concepto de herencia y encapsulamiento.

```python
# Ejemplo básico de clase
class Vehiculo:
    def __init__(self, modelo, año, kilometraje, color="blanco"):
        self.modelo = modelo
        self.año = año
        self.kilometraje = kilometraje
        self.color = color
    def mostrar_propiedades(self):
        print(f"Modelo: {self.modelo}, Año: {self.año}, Km: {self.kilometraje}, Color: {self.color}")

auto1 = Vehiculo("Jetta", 2022, 8000)
auto1.mostrar_propiedades()
```

## 7. Integración: Mini caso práctico (10 min)

* **Contexto:** Eres responsable de crear un sistema sencillo para un parque vehicular de una empresa. Cada auto tiene un modelo, año y kilometraje. Debes identificar:

  * Qué autos deben servicio (más de 10,000 km recorridos).
  * Cuántos autos son anteriores al año 2020.

```python
# Solución ejemplo
flota = [
    Vehiculo("Sentra", 2018, 12000),
    Vehiculo("March", 2021, 9000),
    Vehiculo("Jetta", 2019, 15000),
]

# Autos que requieren servicio y conteo de autos viejos
servicio = []
autos_antes_2020 = 0
for auto in flota:
    if auto.kilometraje > 10000:
        servicio.append(auto.modelo)
    if auto.año < 2020:
        autos_antes_2020 += 1
print("Autos que requieren servicio:", servicio)
print("Cantidad de autos anteriores a 2020:", autos_antes_2020)
```

1. Escribe una función que reciba una lista de números y devuelva la suma de los pares.

```python
def suma_pares(lista):
    return sum([x for x in lista if x % 2 == 0])

print(suma_pares([1, 2, 3, 4, 5, 6]))
```

3. Haz un ciclo que pida números al usuario hasta que escriba "salir" y al final imprima el promedio.

```python
numeros = []
while True:
    entrada = input("Escribe un número (o 'salir'): ")
    if entrada == "salir":
        break
    try:
        numeros.append(float(entrada))
    except:
        print("Dato inválido")
if numeros:
    print("Promedio:", sum(numeros) / len(numeros))
else:
    print("No se ingresaron números.")
```
