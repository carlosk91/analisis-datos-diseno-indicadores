# **Laboratorio de Análisis de Datos Financieros y Diseño de Indicadores**

## **Primera sesión: ¿Qué es un indicador? ¿Qué son los KPIs y OKRs?**

---

**Duración:** 1 hora con 40 minutos  
**Formato:** Jupyter Notebook  
**Objetivo general de la sesión:** Introducir a los estudiantes al concepto de indicadores, Key Performance Indicators (KPIs) y Objectives and Key Results (OKRs), destacando su relevancia para la toma de decisiones empresariales a través del método del caso con datos reales.  

---

## **1. Introducción (15 minutos)**

### **Bienvenida y presentación del curso**
- Explicar la importancia del análisis de datos financieros en la toma de decisiones.  
- Reglas de juego para el laboratorio: Compromisos, guia, proyecto, trabajo colaborativo y uso de herramientas (Jupyter Notebook, ChatGPT Canvas, etc...).
- Mostrar los objetivos del módulo y el producto final esperado: aplicar los conocimientos a un caso real de una empresa.  

---

## **2. Método del caso (50 minutos)**

### **Caso de negocio:**  
**"El Reto del CEO de Panda Delivery"**

### **Contexto del caso**
- Los alumnos ahora son analistas de datos que trabajan para un CEO de una empresa de delivery (similar a Rappi o Uber Eats) llamada Panda Delivery. El CEO ha convocado una reunión de emergencia porque la empresa está perdiendo clientes de forma alarmante. El equipo de estudiantes debe ayudar al CEO a identificar qué está sucediendo y proponer soluciones basadas en la disponibilidad de datos y el análisis de los mismos.

### **Desarrollo del caso**
1. **Presentación del problema (10 min)**: El CEO explica la situación crítica. "Estamos perdiendo clientes de forma masiva y no sabemos por qué. Necesitamos definir qué datos tenemos disponibles y qué información debemos buscar para entender la razón de esta pérdida".  
2. **Discusión guiada (20 min)**: Los estudiantes deben responder:  
    - ¿Qué información creen que necesitan para entender la pérdida de clientes?  
    - ¿Qué métricas e indicadores podrían ayudar a identificar el problema?  
3. **Definición de la estrategia (20 min)**: Los estudiantes deben proponer un plan de acción para definir qué indicadores se deben medir y qué datos se deben recolectar. Los estudiantes deben decidir qué KPI podrían usar para abordar la pérdida de clientes.  
4. **Discusión grupal (10 min)**: Los estudiantes comparten sus hallazgos y recomendaciones. El profesor facilitará la discusión y ofrecerá retroalimentación sobre la selección de indicadores.  

---

## **3. Introducción a los conceptos de KPI y OKR (25 minutos)**

### **Introducción conceptual**
- Definición de "indicador".  
- Introducción a los Key Performance Indicators (KPIs):  
  - Diferencia entre indicadores cuantitativos y cualitativos.
    - Ejemplos de cuantitativos: Ventas totales ($), Ticket promedio, Tasa de churn,
    - Ejemplos de cualitativos: Percepción de la marca, Satisfacción del cliente (NPS)
  - Diferencia entre indicadores adelantados y rezagados. 
    - Ejemplos de rezagados: Ventas
    - Ejemplos de adelantados: Tráfico de la app
  - KPIs -> buenos y malos
  - ¿Qué KPIs creen que serían buenos?, ¿qué KPIs creen que serían malos?
- Explicación de Objectives and Key Results (OKRs) y su relación con la estrategia empresarial. 
- ¿A dónde vamos?, ¿cómo vamos a llegar ahí?
- Los buenos objetivos son ambiciosos
- Diferencia entre OKRs comprometidos y aspiracionales
  - Ejemplos de OKRs comprometidos
  - Ejemplos de OKRs aspiracionales

## ¿Que son los indicadores?

Los indicadores, en un contexto de negocio, son métricas o puntos de datos para cuantificar el desempeño o predecir resultados. Son herramientas importantes para administrar y tomar decisiones, dando una imagen clara del desempeño actual de la compañía comparado con sus objetivos estratégicos. Pueden variar mucho dependiendo del sector, enfoque operativo, y objetivos estratégicos. Pueden medir desempeño financiero, satisfacción del cliente, eficacia operativa, o incluso compromiso de los empleados, incluyendo muchos mas aspectos.

**EJEMPLOS ALUMNOS**


### Ejemplos de como se definen estrategias en entornos empresariales usando indicadores: Objectives and Key Results (OKRs)

- **Key Performance Indicator (KPI):** Los KPIs son valores medibles que demuestran que tan efectiva es una compañia en lograr objetivos de negocio claves. Las organizaciones usan KPIs en multiples niveles para evaluar su éxito en alcanzar objetivos. KPIs de alto nivel pueden enfocarse en el desempeño total de la empresa, mientras que KPIs de bajo nivel pueden enfocarse en procesos de departamentos como ventas, marketing, recursos humanos, soporte y otros.

- **Objectives and Key Results (OKRs):** Una estructura de trabajo para definir y darle seguimiento a objetivos y resultados. Ha sido usado por bastantes compañias de alto perfil para asegurarse de que se están enfocando en los aspectos mas relevantes en toda la organización.
