### **4. Filtrado y operaciones avanzadas (30 minutos)**

**Objetivo:** Filtrar datos utilizando condiciones simples y múltiples, y realizar operaciones de agrupamiento.

#### **Filtrado de datos:**

- Condición simple:

```python
adultos = customers[customers['age'] >= 21]
print(adultos)
```

- Múltiples condiciones:

```python
filtro = customers[(customers['age'] >= 21) & (customers['zip_code'] == 13244)]
print(filtro)
```

#### **Agrupamiento y cálculos:**

- Número de clientes por zona postal:

```python
grouped = customers.groupby('zip_code').agg({'customer_id': 'count'})
print(grouped)
```

- Crear categorías de antigüedad:

```python
customers['registration_date'] = pd.to_datetime(customers['registration_date'])

bins = [
    pd.Timestamp('1900-01-01'),  # Fecha mínima para "Veterano"
    pd.Timestamp('2011-01-01'),  # Fecha mínima para "Intermedio"
    pd.Timestamp('2012-01-01'),  # Fecha mínima para "Nuevo"
    pd.Timestamp('2100-01-01')   # Fecha máxima para "Nuevo"
]

categories = ['Veterano', 'Intermedio', 'Nuevo']

customers['customer_category'] = pd.cut(
    customers['registration_date'], bins=bins, labels=categories, right=False
)

print(customers)
```
