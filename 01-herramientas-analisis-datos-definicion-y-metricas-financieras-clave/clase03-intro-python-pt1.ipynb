## 6. Operaciones con Strings (cadenas de texto) en Python

En Python, los strings permiten almacenar y manipular texto. ¡Son fundamentales para análisis de datos!

### Declarar un string

```python
saludo = "Hola, mundo"
otro = '¡Python es genial!'
```

### Concatenar strings

```python
nombre = "Carlos"
apellido = "Ramírez"
nombre_completo = nombre + " " + apellido
print(nombre_completo)
```

### Repetir strings

```python
print("Hola! " * 3)  # Imprime 'Hola! Hola! Hola! '
```

### Indexación y slicing

```python
texto = "Analítica"
print(texto[0])    # 'A'
print(texto[-1])   # 'a'
print(texto[0:4])  # 'Anal'
print(texto[2:])   # 'alítica'
print(texto[::2])  # 'Aaltc' (cada dos letras)
```

### Longitud de un string

```python
frase = "Indicadores financieros"
print(len(frase))
```

### Métodos útiles de strings

```python
mensaje = "python es increíble"
print(mensaje.upper())       # 'PYTHON ES INCREÍBLE'
print(mensaje.capitalize())  # 'Python es increíble'
print(mensaje.replace("increíble", "fácil"))
print(mensaje.find("es"))   # Devuelve el índice donde aparece 'es'
print(mensaje.split())       # ['python', 'es', 'increíble']
```

### Escape de caracteres especiales

```python
print("Primera línea\nSegunda línea")  # Salto de línea
print("Columna1\tColumna2")            # Tabulador
```

### String formatting (f-strings)

```python
edad = 20
print(f"Hola, mi edad es {edad} años")
```
