# Parte II — Autopsia de la respuesta

Tenemos una respuesta.

Tal vez también tenemos código, gráficas, cálculos y recomendaciones.

Antes de utilizar cualquiera de ellos para tomar una decisión necesitamos responder:

> # ¿Qué evidencia tenemos de que el resultado es correcto?

---

## 1. ¿Qué significa "rentabilidad"?

Nuestra pregunta original decía:

> "¿Está disminuyendo la rentabilidad?"

Pero **rentabilidad** puede significar cosas diferentes.

Por ejemplo:

* margen bruto;
* margen operativo;
* margen neto;
* ROI;
* ROA;
* ROE.

Si nosotros no especificamos cuál queremos utilizar, la IA tendrá que tomar una decisión.

Esa decisión puede ser razonable sin necesariamente ser la correcta para nuestro problema.

---

## Los supuestos invisibles

Cuando no proporcionamos suficiente información, la IA debe completar los espacios.

Por ejemplo:

```text
Analiza los datos y dime por qué está cayendo la rentabilidad.
```

La IA podría decidir que:

```text
rentabilidad = utilidad / ingresos
```

El cálculo podría ser matemáticamente correcto.

Pero quizá el indicador que utiliza la empresa sea:

```text
operating_margin =
(revenue - cogs - operating_expenses) / revenue
```

El problema no necesariamente es que la IA haya cometido un error.

El problema es que:

> **tomó una decisión que nosotros nunca hicimos explícita.**

---

## 2. ¿Revisamos los datos?

Incluso una fórmula correcta puede producir un análisis incorrecto si los datos tienen problemas.

Antes de calcular un indicador podríamos necesitar revisar:

* valores faltantes;
* registros duplicados;
* tipos de datos;
* valores imposibles;
* valores extremos;
* cambios en las unidades;
* periodos incompletos.

Por ejemplo:

````python
df["operating_margin"] = (
    df["revenue"]
    - df["cogs"]
    - df["operating_expenses"]
) / df["revenue"]
````

El código parece razonable.

Incluso puede ejecutarse sin ningún error.

Pero todavía podríamos preguntar:

* ¿existen valores de revenue == 0?
* ¿hay NaN?
* ¿todas las variables utilizan la misma moneda?
* ¿operating_expenses incluye cogs?
* ¿hay registros duplicados?

> **Código que corre ≠ análisis correcto.**

---

## 3. ¿La IA observó algo o lo explicó?

Supongamos que encontramos lo siguiente:

```text
marketing_spend ↑

operating_margin ↓
```

Podemos afirmar que ambas variables cambiaron durante el mismo periodo.

También podríamos encontrar una correlación entre ellas.

Pero sería muy diferente afirmar:

> "El incremento en marketing provocó la caída de la rentabilidad."

Para llegar a esa conclusión necesitamos mucha más evidencia.

---

## Dato, observación, hipótesis y causalidad

Durante un análisis conviene distinguir distintos niveles de afirmación:

| Nivel           | Ejemplo                                                |
| --------------- | ------------------------------------------------------ |
| **Dato**        | Marketing pasó de $100,000 a $125,000                  |
| **Cálculo**     | Marketing aumentó 25%                                  |
| **Observación** | Durante el mismo periodo disminuyó el margen           |
| **Relación**    | Existe una asociación entre ambas variables            |
| **Hipótesis**   | El incremento en marketing podría contribuir al cambio |
| **Causalidad**  | El incremento en marketing provocó la caída del margen |

Cada nivel requiere más evidencia que el anterior.

Una de las tareas más importantes al trabajar con IA será evitar que una **hipótesis plausible** se convierta silenciosamente en una **conclusión**.

---

## 4. Fluidez no significa certeza

Una IA puede escribir:

> "El deterioro del margen probablemente responde a una estrategia agresiva de adquisición de clientes."

Pero debemos preguntar:

> **¿Qué dato demuestra que existe una estrategia agresiva de adquisición?**

Si esa información no aparece en nuestros datos ni en una fuente externa, no tenemos una conclusión.

Tenemos una hipótesis.

# Parte III — Un framework para trabajar con IA

El problema anterior no se resuelve aprendiendo una colección de "prompts mágicos".

Necesitamos especificar mejor el trabajo que queremos realizar.

Durante esta sesión utilizaremos cinco componentes.

| Componente             | Pregunta                               |
| ---------------------- | -------------------------------------- |
| **Objetivo**           | ¿Qué queremos conseguir?               |
| **Contexto**           | ¿Qué necesita saber la IA?             |
| **Restricciones**      | ¿Qué reglas debe respetar?             |
| **Definition of Done** | ¿Qué significa terminar correctamente? |
| **Verificación**       | ¿Cómo comprobaremos que funciona?      |

---

## 1. Objetivo

Define el problema que realmente queremos resolver.

En lugar de:

```text
Analiza los datos.
```

podemos decir:

```text
Determina si el margen operativo se ha deteriorado
durante el periodo disponible.
```

El objetivo define **qué pregunta estamos intentando responder**.

---

## 2. Contexto

Proporciona la información que la IA necesita para interpretar correctamente el problema.

Por ejemplo:

```text
Los datos contienen resultados financieros mensuales.

revenue, cogs y operating_expenses están expresados en MXN.

Para este análisis utilizaremos operating_margin como
principal indicador de rentabilidad.
```

El contexto reduce la necesidad de que el modelo haga suposiciones.

---

## 3. Restricciones

Definen las reglas que queremos que la IA respete.

Por ejemplo:

```text
No elimines registros sin justificarlo.

No inventes variables que no estén disponibles.

No interpretes una correlación como causalidad.

Si no existe suficiente información para responder,
indícalo explícitamente.
```

---

## 4. Definition of Done

Define qué condiciones deben cumplirse antes de considerar terminado el trabajo.

Por ejemplo:

```text
El análisis estará terminado cuando:

- se haya revisado la calidad de los datos;
- se haya calculado el KPI correctamente;
- se haya analizado su evolución;
- se hayan identificado anomalías relevantes;
- el código pueda ejecutarse de principio a fin;
- las conclusiones estén respaldadas por los datos.
```

Esto convierte una instrucción abierta en algo que podemos evaluar.

---

## 5. Verificación

Finalmente definimos cómo comprobaremos el resultado.

Por ejemplo:

```text
Antes de concluir:

- revisa las fórmulas;
- valida manualmente al menos un cálculo;
- busca posibles errores;
- identifica explicaciones alternativas;
- señala qué afirmaciones no pueden demostrarse con los datos.
```

# Parte V — Context Engineering

Hasta ahora hemos definido **qué queremos que haga la IA**.

Existe otra pregunta:

> # ¿Qué necesita saber la IA para poder hacerlo correctamente?

Podemos simplificar:

### Prompt Engineering

> ¿Qué quiero que haga?

### Context Engineering

> ¿Qué necesita saber para hacerlo bien?

---

## Contexto > palabras sofisticadas

Compara:

### Solicitud A

```text
Actúa como un experto financiero de clase mundial.

Analiza nuestros márgenes rigurosamente y produce insights profundos.
```

### Solicitud B

```text
Los datos son mensuales.

Las variables monetarias están expresadas en MXN.

revenue representa ventas netas.

cogs contiene costos directos.

operating_expenses no incluye cogs.

Definimos:

operating_margin =
(revenue - cogs - operating_expenses) / revenue
```

La primera suena más sofisticada.

La segunda contiene más información útil.

---

# ¿Qué contexto puede necesitar una IA?

| Tipo          | Ejemplos                                       |
| ------------- | ---------------------------------------------- |
| **Negocio**   | definiciones, reglas, objetivo                 |
| **Datos**     | variables, unidades, periodos                  |
| **Técnico**   | librerías, entorno, archivos, código existente |
| **Histórico** | decisiones tomadas previamente                 |

---

## No hagas que la IA adivine lo que tú ya sabes

En lugar de:

```text
Analiza esta caída.
```

podemos decir:

```text
La caída corresponde a operating_margin.

Los datos son mensuales.

La empresa realizó un cambio en su estructura de costos en enero.

Queremos comparar el comportamiento posterior al cambio
contra los meses anteriores.
```

Si esa información puede cambiar el análisis, probablemente pertenece al contexto.

---

## El código existente también es contexto

Supongamos que ya tenemos:

```python
def calculate_gross_margin(df):
    return (df["revenue"] - df["cogs"]) / df["revenue"]
```

En lugar de pedir:

```text
Crea una función para operating margin.
```

podemos proporcionar:

```text
Ya existe esta función:

def calculate_gross_margin(df):
    return (df["revenue"] - df["cogs"]) / df["revenue"]

Quiero agregar calculate_operating_margin().

Utiliza:

(revenue - cogs - operating_expenses) / revenue

Mantén el mismo estilo.

No modifiques calculate_gross_margin().
```

Ahora la IA está trabajando dentro del sistema existente.

---

## Las decisiones previas también son contexto

Si ya concluimos:

```text
Los aparentes duplicados representan transacciones diferentes
y deben conservarse.
```

podemos continuar con:

```text
Ya revisamos los posibles duplicados.

Concluimos que representan transacciones diferentes.

No los elimines en los siguientes pasos.
```

Esto evita reabrir decisiones que ya tomamos.

---

## Más contexto no siempre es mejor

Proporcionar todo el repositorio, todas las conversaciones o cientos de páginas de documentación también puede introducir ruido.

Pregunta:

> **¿Esta información podría cambiar la estrategia, el cálculo o la interpretación?**

Si la respuesta es no, probablemente no necesitamos incluirla.

> **Context Engineering no significa darle todo a la IA.**

Significa:

> **darle la información correcta para la decisión que queremos que tome.**

# Parte VI — Vibe Coding: construir con IA

Ya tenemos una especificación.

Ahora necesitamos construir.

Aquí aparecen dos procesos distintos:

## Antes de construir

Especificamos:

```text
Objetivo
   ↓
Contexto
   ↓
Restricciones
   ↓
Definition of Done
   ↓
Verificación
```

## Mientras construimos

Trabajaremos iterativamente:

```text
Understand
   ↓
Plan
   ↓
Implement
   ↓
Run
   ↓
Test
   ↓
Review
   ↓
Iterate
```

Son dos cosas diferentes:

> **El primer framework define el problema.**

> **El segundo define cómo desarrollamos la solución.**

---

## ¿Qué es Vibe Coding?

Podemos utilizar lenguaje natural para dirigir gran parte del proceso de programación.

La IA puede ayudarnos a:

* entender código;
* proponer cambios;
* escribir funciones;
* ejecutar análisis;
* encontrar errores;
* generar pruebas;
* modificar soluciones.

El riesgo es producir código más rápido de lo que somos capaces de entenderlo o verificarlo.

---

## Trabaja en cambios pequeños

En lugar de:

```text
Limpia los datos, calcula todos los KPIs, crea gráficas,
analiza correlaciones y dame recomendaciones.
```

podemos trabajar así:

```text
Primero revisa la calidad de los datos.

No modifiques nada todavía.
```

Después:

```text
Propón qué harías con los problemas encontrados.

No hagas cambios todavía.
```

Y finalmente:

```text
Implementa únicamente las decisiones que acabamos de aprobar.
```

Antes de aceptar algo producido con IA, pregúntate:

* ¿entendió el problema?
* ¿tenía el contexto necesario?
* ¿ejecutamos lo que produjo?
* ¿lo verificamos?
* ¿podemos explicar nosotros mismos el resultado?

> **La IA puede producir el trabajo.**

> **Tú sigues siendo responsable del resultado.**
