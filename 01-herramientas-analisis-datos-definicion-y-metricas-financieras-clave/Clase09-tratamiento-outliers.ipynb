#### **1.2 Técnicas de identificación**

**Detección de outliers univariados**

Se enfoca en el análisis de outliers de una sola variable a la vez.
  
  - **Técnicas visuales**

    - **<ins>Histograma:</ins>** Visualización gráfica de la frecuencia de los valores en un conjunto de datos. Examinar su patrón puede darnos indicios sobre la existencia de valores atípicos.
      <center>
         <img src="https://fernandoblancopsy.com/wp-content/uploads/2019/11/image-19.png?w=500"  width="400" />
      </center> 

    <br>

    - **<ins>Boxplot:</ins>** Visualizaciones gráficas que muestran la distribución de los datos, incluyendo la mediana, los cuartiles y el rango de los datos.

      - **Caja**: Representa el rango intercuartílico (IQR), que es la distancia entre el primer y el tercer cuartil.
      - **Línea de la mediana**: Divide el conjunto de datos en dos mitades iguales.
      - **Bigotes**: Son líneas que se extienden desde los extremos de la caja hasta los valores mínimo y máximo que están dentro de 1.5 veces el IQR desde el primer y tercer cuartil.
      - **Valores atípicos (outliers)**: Son los valores que se encuentran fuera de estos límites y se consideran atípicos.

      <center>
         <img src="https://miro.medium.com/v2/resize:fit:828/format:webp/0*BenFO7Hi0cLJA7_u"  width="500" />
      </center> 


  - **Técnicas estadísticas**
    - **<ins>Z-Score:</ins>** Medida estadística que indica cuantas desviaciones estándar un dato está por encima o por debajo de la media. 
    
      Generalmente, se considera que un Z-score absoluto mayor a 3 identifica un outlier. Este umbral se basa en la regla empírica, que establece que el 99.7% de los datos en una distribución normal se encuentran dentro de tres desviaciones estándar de la media. 
    
      Por lo tanto, el método del Z-score es adecuado principalmente para distribuciones normales.

      <center>
          <img src="https://miro.medium.com/v2/resize:fit:828/format:webp/0*aRgoMzeHDoHBL4Wk"  width="380" />
      </center> 

    <br>

    - **<ins>Rango intercuartílico (IQR):</ins>** Mide la dispersión entre el primer cuartil (Q1) y el tercer cuartil (Q3) de una distribución. 
    
      Se utiliza para identificar valores atípicos definiendo límites fuera de los cuales los datos se consideran outliers. Estos límites se calculan así:

      - Límite inferior = Q1 - (k * IQR)
      - Límite superior = Q3 + (k * IQR)

      Donde Q1 es el primer cuartil y Q3 es el tercer cuartil. El valor común para k es 1.5, aunque un k de 3 o más puede identificar outliers extremos. Los valores fuera de estos límites se consideran atípicos

### **2. "Elite Properties" Valores outliers (35 minutos)**

**Ejercicio Guiado:**

**Caso:** *Elite Properties*, una empresa inmobiliaria australiana, identificó que en el último año la venta de casas en el área de Melbourne ha experimentado un auge.

Melbourne se encuentra en el estado de Victoria. La empresa recopiló, en el último trimestre de 2024, el precio de las casas en cinco áreas de Victoria que comparten comportamientos macroeconómicos similares.

El equipo de análisis de la empresa utilizará esta información para evaluar la conveniencia de invertir en alguna área específica.

<center>
        <img src="https://img.freepik.com/premium-vector/people-buying-house-concept-man-woman-with-magnifying-glass-evaluate-real-estate-private-property-savings-family-budget-economics-investment-cartoon-flat-vector-illustration_118813-15861.jpg?semt=ais_hybrid"  width="250" />
      </center> 

  **Registros:** 
  
  Se tiene la información de 2,639 viviendas.

  **Variables**

  - *Área:* Área dentro del estado de Victoria donde se encuentra la casa.
  - *Precio:* En dólares.

  **Propuesta**

  Se propone identificar los valores outlieres en precio de las casas y analizar su existencia en las diferentes 5 áreas.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

pd.set_option('display.float_format', '{:,.2f}'.format)

houses = pd.read_csv('data/houses_data_outliers.csv')
```

```python
houses.info()
```

```python
houses['Precio'].describe()
```

Analizar el precio de las casa por área.

```python
precio_area = houses.groupby('Area').agg(
    Casas=('id', 'count'),
    Precio_Mediana=('Precio', 'median'),
    Precio_Promedio=('Precio', 'mean'),
    Precio_Maximo=('Precio', 'max'),
    Precio_Minimo=('Precio', 'min')
).reset_index()

precio_area
```

**Histograma**

Al analizar el histograma, se observa que los precios altos tienen una baja frecuencia.

```python
plt.hist(houses['Precio'], bins=15, edgecolor='darkgray',color='green')
# Añadir títulos y etiquetas
plt.title('Histograma Precio Casas')
plt.xlabel('Precio (USD)')
plt.ylabel('Frecuencia')
plt.show()
```

**Box-Plot**

Los puntos representan datos outliers.

```python
plt.boxplot(houses['Precio'],boxprops=dict(color='green'),medianprops=dict(color='gray'))
# Añadir títulos y etiquetas
plt.title('Boxplot Precio Casas')
plt.ylabel('Precio (USD)')
plt.show()
```

**Z-Score**

Se identifican los outliers bajo la metodología *z-score*.

```python
# Calcular z-score para cada punto y su valor absoluto
z_scores = zscore(houses['Precio'])
abs_z_scores = np.abs(z_scores)

# Seleccionar outliers con un threshold de 3
houses['outlier_zscore']=(abs_z_scores > 3)*1
houses.head()
```

Detalle outliers por área

```python
zscore_group=houses.groupby('Area').agg(
    Casas=('id', 'count'),
    Outlier=('outlier_zscore', 'sum')
).reset_index()

zscore_group['Ratio_out']=zscore_group['Outlier']/zscore_group['Casas']

zscore_group
```

Precio promedio no outliers y outliers.

```python
houses.groupby(['Area','outlier_zscore']).agg(
    Casas=('id', 'count'),
    Precio_Promedio=('Precio', 'mean')
).reset_index()
```

**Rango intercuartílico (IQR)**

Se identifican los outliers bajo la metodología *IQR*.

```python
# Calcular percentiles
q75 = houses['Precio'].quantile(0.75)
q25 = houses['Precio'].quantile(0.25)

# Obtener IQR
iqr = q75 - q25

# thresholds
upper = q75 + (1.5 * iqr)
lower = q25 - (1.5 * iqr)

houses['outlier_iqr']=((houses['Precio'] < lower) | (houses['Precio'] > upper))*1
houses.head()
```

Detalle outliers por área

```python
iqr_group=houses.groupby('Area').agg(
    Casas=('id', 'count'),
    Outlier=('outlier_iqr', 'sum'),
).reset_index()

iqr_group['Ratio_out']=iqr_group['Outlier']/iqr_group['Casas']

iqr_group
```

Precio promedio no outliers y outliers.

```python

houses.groupby(['Area','outlier_iqr']).agg(
    Casas=('id', 'count'),
    Precio_Medio=('Precio', 'mean')
).reset_index()
```

**Rango intercuartílico (IQR)**

**Ejercicios para los estudiantes:**

Identificar los **datos extremos** bajo la metodología de IQR

```python
# Calcular percentiles
q75 = houses['Precio'].quantile(0.75)
q25 = houses['Precio'].quantile(0.25)

# Obtener IQR
iqr = q75 - q25

# thresholds (3 Datos Extremos)
upper = q75 + (3 * iqr)
lower = q25 - (3 * iqr)

houses['extremo_iqr']=((houses['Precio'] < lower) | (houses['Precio'] > upper))*1
houses.head()
```

Detalle extremos por área

```python
iqr_group=houses.groupby('Area').agg(
    Casas=('id', 'count'),
    Extremo=('extremo_iqr', 'sum'),
).reset_index()

iqr_group['Ratio_ext']=iqr_group['Extremo']/iqr_group['Casas']

iqr_group
```

Precio promedio no extremos y extremos.

```python
houses.groupby(['Area','extremo_iqr']).agg(
    Casas=('id', 'count'),
    Precio_Medio=('Precio', 'mean')
).reset_index()
```
