## 2. Tuplas en Python

Una **tupla** es una colección ordenada e inmutable. Ideal para datos que no deben cambiar.

```python
# Crear una tupla
indicador = ("KPI_Liquidez", 2024, 1.15)
print(indicador)

# Acceso por índice
print(indicador[0])  # 'KPI_Liquidez'
print(indicador[-1]) # 1.15

# Slicing
otra = indicador + ("Rentabilidad", 2022)
print(otra[1:4])

# Tuplas anidadas
anidada = (1, 2, ("ROI", "ROE"), (3,4), ("Liquidez",(1,2)))
print(anidada[2][1])  # 'ROE'
```

## 3. Listas en Python

Las **listas** son colecciones ordenadas y mutables. Muy utilizadas en análisis de datos.

```python
# Crear y acceder a listas
empresas = ["Cemex", "Bimbo", "América Móvil"]
print(empresas[1])  # 'Bimbo'

# Añadir y extender
empresas.append("Femsa")
empresas.extend(["Alsea", "Walmart"])
print(empresas)

# Cambiar un valor
empresas[0] = "Grupo México"

# Eliminar elementos
del empresas[2]
print(empresas)

# Slicing
print(empresas[1:3])

# Listas anidadas
datos = [["Empresa", "Ingresos"], ["Cemex", 15000], ["Bimbo", 13000]]
print(datos[1][1])  # 15000
```

## 4. Conjuntos en Python

Los **conjuntos (sets)** son colecciones desordenadas de elementos únicos. Útiles para eliminar duplicados o realizar operaciones lógicas.

```python
# Crear un set
empresas_set = {"Cemex", "Bimbo", "Bimbo", "Alsea"}
print(empresas_set)  # {'Bimbo', 'Cemex', 'Alsea'}

# Añadir/eliminar
empresas_set.add("Walmart")
empresas_set.remove("Alsea")

# Verificar si existe
print("Cemex" in empresas_set)

# Operaciones de conjuntos
set1 = {"Cemex", "Bimbo", "Femsa"}
set2 = {"Bimbo", "Walmart", "Alsea"}
print(set1 | set2)   # Unión
print(set1 & set2)   # Intersección
print(set1 - set2)   # Diferencia
```

## 5. Diccionarios en Python

Un **diccionario** almacena pares clave-valor. Ideal para información estructurada.

```python
# Crear un diccionario
indicadores = {"Liquidez": 1.15, "Rentabilidad": 0.12, "Apalancamiento": 0.65}
print(indicadores)

# Acceso y modificación
print(indicadores["Liquidez"])
indicadores["Solvencia"] = 0.89

del indicadores["Apalancamiento"]

# Claves y valores
print(indicadores.keys())
print(indicadores.values())
```
