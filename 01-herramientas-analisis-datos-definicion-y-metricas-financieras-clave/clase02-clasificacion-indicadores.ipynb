# **Sesión 2: Entendiendo y clasificando indicadores**

## **Objetivo:**
Proporcionar a los estudiantes un entendimiento práctico de los diferentes tipos de indicadores clave de rendimiento (KPIs) y su clasificación, así como los conceptos básicos de unit economics. La sesión se enfocará en discutir conceptos clave y resolver preguntas del caso sin depender de programación compleja.

---

## **Estructura de la sesión:**

### **1. Introducción a los KPIs (20 minutos)**
- Breve explicación sobre qué son los KPIs y su importancia para las empresas.
- Clasificación básica de KPIs:
  - **Cuantitativos vs. Cualitativos:**
    - Cuantitativos: Medibles y expresados en números (e.g., Ingreso, Margen de Ganancia).
    - Cualitativos: Más subjetivos y basados en percepción (e.g., Satisfacción del Cliente, NPS).
  - **Estratégicos, Operacionales y Funcionales:**
    - Estratégicos: Metas de alto nivel alineadas con la visión empresarial.
    - Operacionales: Enfocados en la eficiencia diaria.
    - Funcionales: Relacionados con departamentos específicos.

#### **Actividad inicial:**
**Caso:** Eres parte del equipo de estrategia de una startup de tecnología. El CEO plantea las siguientes preguntas para entender el desempeño del negocio:
1. ¿Qué métricas indicarían que estamos creciendo como empresa?
2. ¿Cómo medirías la satisfacción de los clientes?
3. ¿Qué KPI recomendarías para evaluar la eficiencia de las operaciones diarias?

**Discusión grupal:**
- Los estudiantes proponen ejemplos de KPIs para cada categoría.
- El instructor guía la conversación y proporciona retroalimentación.

---

### **2. Unit Economics y su importancia (30 minutos)**
- **Conceptos clave:**
  - **Costo de Adquisición de Clientes (CAC):**
    - Fórmula: $CAC = \frac{\text{Gastos en Marketing y Ventas}}{\text{Número de Clientes Nuevos}}$
    - Indicador de cuánto cuesta adquirir un cliente.
  - **Valor de Vida del Cliente (LTV) a 24 meses:**
    - Fórmula: $LTV = \frac{\text{Ingresos Totales de los Primeros 24 Meses de la vida de los clientes}}{\text{Clientes con 24 meses de tenure}}$
    - Mide el valor que un cliente genera durante sus primeros 24 meses con la empresa.
  - **Margen Bruto:**
    - Fórmula: $\text{Margen Bruto} = \frac{\text{Ingresos Totales} - \text{Costos Directos}}{\text{Ingresos Totales}}$
    - Indica la rentabilidad después de cubrir los costos directos.

#### **Actividad guiada:**
**Caso:** Una startup de delivery está preocupada por su sostenibilidad financiera. El CEO plantea las siguientes preguntas:
1. ¿Estamos gastando más en adquirir clientes de lo que generamos con ellos? Por ejemplo, si el CAC es 50MXN y el LTV es 100MXN, ¿cómo interpretarías esta relación para evaluar la sostenibilidad del modelo de negocio? (Comparar CAC y LTV).
2. ¿Qué porcentaje de nuestros ingresos estamos destinando a costos directos? Por ejemplo, si el margen bruto es del 60%, esto implicaría que el 40% de los ingresos se destina a cubrir costos directos.
3. ¿Cómo podríamos mejorar nuestras métricas unitarias para aumentar la rentabilidad?

**Discusión grupal:**
- Los estudiantes analizan las preguntas y proponen estrategias basadas en los conceptos explicados.

---

### **3. Clasificación práctica de indicadores (30 minutos)**
- **Ejemplos por categoría:**
    - Cuantitativos: Ingreso, Margen de Ganancia, LTV.
    - Cualitativos: NPS, CSAT.
    - Estratégicos: ROI, Participación de Mercado.
    - Operacionales: Tiempo de Entrega, Rotación de Inventario. Estos indicadores evalúan la eficiencia diaria de las operaciones generales de la empresa, como logística o gestión de inventarios.
    - Funcionales: Tasa de Rotación (Recursos Humanos), CTR (Marketing). Estos indicadores están enfocados en el desempeño y eficiencia de departamentos específicos, como el éxito de campañas de marketing o la estabilidad laboral.

#### **Actividad de caso práctico:**
**Caso:** Una empresa de e-commerce quiere definir indicadores para diferentes áreas del negocio. Responde las siguientes preguntas:
1. ¿Qué KPI cuantitativo usarías para medir el crecimiento financiero?
2. ¿Qué KPI cualitativo recomendarías para evaluar la experiencia del cliente?
3. ¿Qué KPI operacional sería relevante para medir la eficiencia logística?
4. ¿Qué KPI funcional recomendarías para marketing y recursos humanos?

**Discusión final:**
- Los estudiantes comparten sus respuestas y discuten cómo estos indicadores pueden ayudar a alcanzar los objetivos empresariales.

---

### **4. Conclusión y cierre (20 minutos)**
- Repaso de los conceptos aprendidos:
  - Importancia de clasificar indicadores según las necesidades del negocio.
  - Relación entre KPIs estratégicos, operativos y funcionales.
  - Uso de unit economics para evaluar la sostenibilidad de un modelo de negocio.
- **Reflexión grupal:**
  - ¿Cómo pueden aplicar lo aprendido en futuros proyectos?
  - ¿Qué preguntas adicionales tienen sobre el uso de indicadores en la toma de decisiones?

---

### **Resultados esperados:**
- Comprender los diferentes tipos de KPIs y su clasificación.
- Comprender indicadores relevantes para resolver problemas de negocio.
- Comprender los conceptos básicos de unit economics y cómo aplicarlos en análisis estratégicos.
