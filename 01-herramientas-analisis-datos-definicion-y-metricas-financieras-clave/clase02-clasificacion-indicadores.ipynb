# **Sesión 2: Entendiendo y clasificando indicadores**

## **Objetivo:**
Proporcionar a los estudiantes un entendimiento práctico de los diferentes tipos de indicadores clave de rendimiento (KPIs) y su clasificación, así como los conceptos básicos de unit economics. La sesión se enfocará en discutir conceptos clave y resolver preguntas del caso sin depender de programación compleja.

---

## **Estructura de la sesión:**

### **1. Introducción a los KPIs (20 minutos)**
- Breve explicación sobre qué son los KPIs y su importancia para las empresas.

  - Métrica que indica **DESEMPEÑO**. 
  - ¿Cómo es distinto a una simple métrica?
  - **¿Por qué es importante un KPI?**
    - **Monitoreo de progreso**
    - **Identificación de oportunidades de mejora**
    - **Facilita la toma de decisiones**

  - ¿Cómo deben ser?, ¿por qué? Recuerden **SMART** para que sea **EFECTIVO**
    - Specific
    - Measurable
    - Achievable
    - Relevant
    - Time-bound
- Clasificación básica de KPIs:
  - **Cuantitativos vs. Cualitativos:**
    - Cuantitativos: Medibles y expresados en números (e.g., Ingreso, Margen de Ganancia, Ventas Totales).
    - Cualitativos: Más subjetivos y basados en percepción (e.g., Satisfacción del Cliente, NPS, Reputación de Marca).
  - **Estratégicos, Operacionales y Funcionales:**
    - Estratégicos: Metas de alto nivel alineadas con la visión empresarial. (e.g., Market Share, Crecimiento MoM, Metas de ventas)
    - Operacionales: Enfocados en la eficiencia diaria. (e.g., Tiempos de respuesta al cliente, Tiempos de producción, Avg. ETA en un Uber)
    - Funcionales: Relacionados con departamentos específicos. (e.g. HR: Tasa de rotación de personal, employee engagement, Mkt: CAC, Growth)

#### **Actividad inicial:**
**Caso:** Eres creador de contenido en YouTube. Quieres crear KPIs para entender el _performance_ de tu canal:
1. ¿Qué KPIs indicarían que está creciendo tu canal?
  - Número de suscriptores 
  - Views / Impresiones
  - Watch Time per Video
2. ¿Cómo medirías la satisfacción de tus "seguidores"?
  - Tasa de Me Gusta
  - Comentarios positivos
  - Redes sociales (comentarios)
3. ¿Qué métricas medirías para saber qué contenido priorizar?
  - Tasa de retención de audiencia (Minutos de video visto / Tiempo total de video)
  - Engagement Rate (Likes + Shares + Comentarios / Total Views) -> ¿les pondrían un peso distinto a cada uno?
  - Tasa de crecimiento de suscriptores por tipo de contenido (esta es más difícil de medir)
4. ¿Qué KPIs recomendarías para saber qué es más monetizable?
  - Revenue per Video
  - CPM (Cost per Million)
  - Tasa de Conversión a otros productos
  - ESTO **DEPENDE** DE TU BUSINESS MODEL -> SIEMPRE TENERLO EN CUENTA

Si quieres emprender un canal, ¿qué considerarías para que KPIs para que fuera profitable?

### **2. Unit Economics y su importancia (30 minutos)**

- Nos vamos a concentrar en un **costo unitario**—¿cuánto gano o pierdo por cada venta o cada cliente?
- ¿Por qué son importantes los **unit economics**?
- **Conceptos clave:**
  - **Costo de Adquisición de Clientes (CAC):**
    - Fórmula: $CAC = \frac{\text{Gastos en Marketing y Ventas}}{\text{Número de Clientes Nuevos}}$
    - Indicador de cuánto cuesta adquirir un cliente.
    - Ejemplo: Si te cuesta $10,000MXN adquirir 200 clientes, su CAC es de $50MXN
  - **Valor de Vida del Cliente (LTV) a 24 meses:**
    - Fórmula: $LTV = \frac{\text{Ingresos Totales de los Primeros 24 Meses de la vida de los clientes}}{\text{Clientes con 24 meses de tenure}}$
    - Mide el valor que un cliente genera durante sus primeros 24 meses con la empresa.
    - Ejemplo: Si en 24 meses, 100 clientes te generan $10,000MXN su LTV24 es de $100MXN pero **NO CONSIDERA OTROS COSTOS**
  - **Margen Bruto:**
    - Fórmula: $\text{Margen Bruto} = \frac{\text{Ingresos Totales} - \text{Costos Directos}}{\text{Ingresos Totales}}$
    - Indica la rentabilidad después de cubrir los costos directos.
    - Ejemplo: Si vendes $10,000MXN y tus costos de producción son de $6,000MXN tu margen bruto es de 40%

#### **Actividad guiada:**
**Caso:** Una startup de delivery está preocupada por su sostenibilidad financiera. El CEO plantea las siguientes preguntas:
1. ¿Estamos gastando más en adquirir clientes de lo que generamos con ellos? Por ejemplo, si el CAC es 50MXN y el LTV es 100MXN, ¿cómo interpretarías esta relación para evaluar la sostenibilidad del modelo de negocio? (Comparar CAC y LTV).
  - **Primero**, si el LTV es a 24 meses, considerar todo el tiempo que tardamos en recuperar ese dinero. 
  - **Segundo**, no sabemos a cuánto ascienden otros costos. 
2. ¿Qué porcentaje de nuestros ingresos estamos destinando a costos directos? 
  - Si el margen bruto es del 60%, esto implicaría que el 40% de los ingresos se destina a cubrir costos directos.
3. ¿Cómo podríamos mejorar nuestras métricas unitarias para aumentar la rentabilidad?
  - Reducir CAC (marketing más efectivo, programas de referidos).
  - Aumentar LTV (retención, ventas cruzadas, mejoras en el servicio).
  - Mejorar Margen Bruto (negociar con proveedores, optimizar procesos).
4. ¿Hay algo que no está viendo el CEO?
  - Costos ocultos (por ejemplo, indemnizaciones, costos de tecnología, mantenimiento).
  - Tiempo de recuperación: No es lo mismo recuperar el costo de adquisición en 1 mes que en 2 años.
  - Estrategias de crecimiento que afecten la rentabilidad a corto plazo pero beneficien a largo plazo.

**Discusión grupal:**
- Los estudiantes analizan las preguntas y proponen estrategias basadas en los conceptos explicados.

### **3. Clasificación práctica de indicadores (30 minutos)**
    - Cuantitativos
    - Cualitativos
    - Estratégicos
    - Operacionales
    - Funcional
- **Ejemplos por categoría:**
    - Ingreso                               -> Cuantitativo
    - Margen de Ganancia                    -> Cuantitativo, Estratégico
    - Tasa de Productos Defectuosos         -> Cuantitativo, Operacional
    - LTV                                   -> Cuantitativo
    - CTR (Click Through Rate)              -> Cuantitativo, Funcional
    - Retorno sobre la Inversión (ROI)      -> Cuantitativo, Estratégico
    - Costo por Lead                        -> Cuantitativo, Funcional
    - On-time Delivery Rate                 -> Cuantitativo, Operacional
    - Tasa de Conversión                    -> Cuantitativo, Funcional
    - Participación de Mercado              -> Cuantitativo, Estratégico
    - Employee Engagement                   -> Cuantitativo, Funcional
    - Tiempo de Entrega de Pedidos          -> Cuantitativo, Operacional
    

#### **Actividad de caso práctico:**
**Caso:** Una empresa de e-commerce quiere definir indicadores para diferentes áreas del negocio. Responde las siguientes preguntas:
1. ¿Qué KPI cuantitativo usarías para medir el crecimiento financiero?
  - Ventas Mensuales, Ingresos Promedio por Pedido, Tasa de Crecimiento de Ventas.
2. ¿Qué KPI cualitativo recomendarías para evaluar la experiencia del cliente?
  - NPS (Net Promoter Score), CSAT (Customer Satisfaction Score), comentarios en redes sociales o encuestas post-compra.
3. ¿Qué KPI operacional sería relevante para medir la eficiencia logística?
  - Tiempo de Entrega Promedio, Tasa de Entregas a Tiempo, Índice de Pedidos Retrasados.
4. ¿Qué KPI funcional recomendarías para marketing y recursos humanos?
  - **Marketing:** CTR, Tasa de Conversión en anuncios, Costo por Lead.
  - **Recursos Humanos:** Tasa de Rotación de Personal, Índice de Satisfacción de Empleados, Tasa de Absentismo.

**Discusión final:**
- Los estudiantes comparten sus respuestas y discuten cómo estos indicadores pueden ayudar a alcanzar los objetivos empresariales.

  - Importancia de clasificar indicadores según las **necesidades del negocio**.
  - Relación entre KPIs estratégicos, operativos y funcionales.
  - Uso de unit economics para evaluar la sostenibilidad de un modelo de negocio.
- **Reflexión grupal:**
  - ¿Cómo pueden aplicar lo aprendido en futuros proyectos?
  - ¿Qué preguntas adicionales tienen sobre el uso de indicadores en la toma de decisiones?

- Comprender indicadores relevantes para resolver problemas de negocio.
