#### **Pregunta 2**

La empresa Alta Tech pide que calcules el margen bruto para tomar decisiones sobre su desempeño:

  - **Margen Bruto:**
    - Fórmula: $\text{Margen Bruto} = \frac{\text{Ingresos Totales} - \text{Costos Directos}}{\text{Ingresos Totales}}$


```python
import pandas as pd

ventas = pd.read_excel('data/alfa_tech_quizz1.xlsx', sheet_name= 'Ventas')
gastos = pd.read_excel('data/alfa_tech_quizz1.xlsx', sheet_name= 'Gastos')
```

- **¿Cuál fue el margen bruto promedio de Alfa Tech en el año de 2023 y 2024?**

```python
#Merge data
merged = ventas.merge(gastos, on="Fecha", how="left")
merged["margen_bruto"] = (merged["Ventas"] - merged["Gastos"]) / merged["Ventas"]
```

```python
#Formato Fecha
#Columna de año
merged['Year'] = merged['Fecha'].dt.year
```

```python
merged.groupby('Year')['margen_bruto'].mean()
```

- **¿Cuál fue el margen bruto por cada uno de los meses del 2024?**

```python
merged[merged['Fecha']>'2023-12-01'][['Fecha','margen_bruto']]
```
