# **Sesión 7: ChatGPT para prompt engineering**

**Objetivo:** Desarrollar habilidades en prompt engineering utilizando ChatGPT para asistir en la creación, interpretación y automatización de indicadores financieros clave (KPIs), facilitando el análisis de desempeño empresarial a través de la formulación de instrucciones precisas que permitan generar reportes, explicar métricas financieras y optimizar la toma de decisiones con base en datos.
