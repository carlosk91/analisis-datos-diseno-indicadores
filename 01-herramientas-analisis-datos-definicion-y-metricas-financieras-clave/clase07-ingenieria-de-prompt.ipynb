### **1.Cómo hablarle a ChatGPT: estructura de un prompt**

  - Instrucción: Es el núcleo del prompt, donde defines claramente lo que quieres que el modelo haga. Debe ser directo y específico.

    Ejemplo: «Explica», «Resume», «Traduce», «Genera una lista», «Actúa como un experto en…»
  
  - Contexto: Aquí defines el escenario, la situación o los detalles adicionales que ayudan al modelo a entender mejor la tarea. El contexto puede incluir el propósito, el público objetivo o cualquier información relevante.

    Ejemplo: «Estoy escribiendo un artículo para principiantes sobre inteligencia artificial»

  - Formato: Indica cómo debe estructurarse la respuesta. Puedes pedir una lista, un párrafo, un código, una tabla, etc.

    Ejemplo: «Organiza la información en una tabla de 5 columnas», «Escribe la respuesta en formato JSON».»

  - Input: Es la información o el dato que proporcionas al modelo para que trabaje sobre ello. Puede ser un texto, una pregunta, un concepto o cualquier dato relevante.

    Ejemplo: Si pides una traducción, el input sería el texto que quieres traducir. («Fear of a name only increases fear of the thing itself.»)

  - Output: Especifica qué tipo de resultado esperas. Puedes definir la extensión, el nivel de detalle o el enfoque que debe tener la respuesta.

    Ejemplo: «Genera un resumen de 100 palabras», «Escribe una respuesta técnica y detallada».

  - Tono: Define el estilo o la actitud que debe tener la respuesta. Puede ser formal, informal, técnico, amigable, persuasivo, etc.

    Ejemplo: «Escribe en un tono profesional», «Usa un lenguaje sencillo y amigable».

### 4. Zero-shot, One-shot, Few-shot Prompting

Las técnicas **Zero-shot**, **One-shot** y **Few-shot** se utilizan en distintos modelos de inteligencia artificial y no se limitan únicamente a los modelos de generación de texto.

---

#### • Zero-shot

Se refiere a la capacidad de un modelo para realizar una tarea sin ejemplos previos, únicamente a partir de una breve descripción de la tarea.

**Ejemplo:**
“Diseña 10 preguntas de deletreo para un niño de 7 años.”

---

#### • One-shot

Se refiere a la capacidad de un modelo para realizar una tarea después de haber recibido **un solo ejemplo** que sirva de guía.

**Ejemplo:**
“Diseña 10 preguntas de deletreo para un niño de 7 años. Sigue el siguiente ejemplo:
¿*Cómo se escribe la palabra del animal que ladra?*”

---

#### • Few-shot

Se refiere a la capacidad de un modelo para realizar una tarea después de haber recibido **varios ejemplos** que lo ayuden a comprender el contexto.

**Ejemplo:**
“Diseña 10 preguntas de deletreo para un niño de 7 años. Sigue los siguientes ejemplos:
¿*Cómo se escribe la palabra del animal que ladra?*
¿*Cómo se escribe la fruta dulce y roja que crece en los árboles?*”

---

Por defecto, ChatGPT está diseñado para no requerir el uso de One-shot o Few-shot prompting, ya que está entrenado con una base de conocimientos muy amplia. Sin embargo, estas técnicas pueden mejorar la **especificidad** y **precisión** de los prompts cuando se busca un resultado más controlado.

### **6.Optimizar resultados con estructuras de texto específicas**

1. Tablas
Las tablas son una forma efectiva de presentar información de manera clara y concisa. Para pedir a ChatGPT que genere una tabla, es necesario proporcionarle una indicación clara y específica sobre el contenido y la estructura de la tabla. Por ejemplo:

Genera una tabla con los nombres, edades y ciudades de los 10 estudiantes con mejor promedio en la escuela.


2. CSV
Los archivos CSV (Comma Separated Values) son una forma común de almacenar datos de manera estructurada. Para pedir a ChatGPT que genere un archivo CSV, es necesario proporcionarle una indicación clara sobre los datos que se deben incluir en el archivo. Por ejemplo:

Genera un archivo CSV con los nombres, edades y correos electrónicos de los 20 empleados de la empresa.


3. Mapa mental
Los mapas mentales son una forma visual de presentar información de manera organizada. Para pedir a ChatGPT que genere un mapa mental, es necesario proporcionarle una indicación clara sobre los temas y la estructura del mapa. Por ejemplo:

Genera un mapa mental que describa los diferentes elementos que componen la teoría de la evolución de Darwin.


4. Outline
Los outlines son una forma efectiva de presentar información de manera jerárquica y estructurada. Para pedir a ChatGPT que genere un outline, es necesario proporcionarle una indicación clara sobre la estructura y los temas a incluir en el outline. Por ejemplo:

Genera un outline sobre los diferentes tipos de energía renovable, incluyendo la energía solar, la energía eólica y la energía hidráulica.


5. JSON
El formato JSON (JavaScript Object Notation) es una forma común de intercambio de datos estructurados. Para pedir a ChatGPT que genere un archivo JSON, es necesario proporcionarle una indicación clara sobre los datos que se deben incluir en el archivo y la estructura del mismo

### **7.Mejorar resultados iterando**

Mejorar resultados iterando:
ㅤ
La mejora de resultados a través de iteraciones, consiste en generar respuestas iniciales que se van refinando y mejorando a través de una serie de iteraciones. En cada iteración, se toma como punto de partida la respuesta anterior y se le hace una serie de ajustes, adiciones o eliminaciones para mejorar su calidad, dando un feedback al modelo sobre qué se debe mantener y qué se debe cambiar.
ㅤ
Este proceso de refinamiento continuo permite que las respuestas generadas sean cada vez más precisas, relevantes y satisfactorias según nuestras necesidades, y se logra a través de la retroalimentación constante y el ajuste incremental. Esta técnica puede ser útil en diferentes casos:
ㅤ
    • Mejorar una misma respuesta hasta lograr el resultado deseado, dándole feedback a ChatGPT sobre qué debe cambiar en cada respuesta.
    • Generar un elemento complejo o de varias etapas, dividiéndolo en varias subtareas. Por ejemplo.
    • 🧑‍💻 Generar código en diferentes lenguajes de programación, añadiendo elementos o funciones sobre cada respuesta generada.



🔁 Iteración 1 – Prompt inicial:
📝 Prompt:

Explica qué es el ROI con tus propias palabras.

📥 Respuesta del modelo:

El ROI (Return on Investment) es una métrica financiera que mide cuánto dinero se ha ganado en relación con el dinero que se ha invertido.

🗣 Feedback:
"Está bien, pero es muy general. Añade una fórmula y un ejemplo numérico simple."

🔁 Iteración 2 – Prompt ajustado:
Explica qué es el ROI incluyendo su fórmula y un ejemplo numérico sencillo.

📥 Respuesta del modelo:

El ROI (Return on Investment) es una métrica que calcula el rendimiento de una inversión.
Fórmula: ROI = (Ganancia - Inversión) / Inversión
Ejemplo: Si inviertes $1000 y obtienes $1200, el ROI es (1200-1000)/1000 = 0.2 o 20%.

🗣 Feedback:
"Ahora agrégale una analogía sencilla para que un estudiante de preparatoria lo entienda mejor."

🔁 Iteración 3 – Prompt mejorado:
Explica qué es el ROI con fórmula, ejemplo numérico y una analogía que sea fácil de entender para un estudiante de preparatoria.

📥 Respuesta del modelo:

El ROI (Return on Investment) mide cuánto ganaste en comparación con lo que invertiste.
Fórmula: ROI = (Ganancia - Inversión) / Inversión
Ejemplo: Si inviertes $1000 y obtienes $1200, el ROI es 20%.
Analogía: Es como si compras ingredientes por $100 para hacer pasteles y luego los vendes por $120. Ganaste $20 sobre tu inversión, lo que representa un 20% de ganancia.

### 8

Role play: pedirle a ChatGPT que actúe de cierta manera
ChatGPT es capaz de actuar como si fuera una persona específica o de adoptar una personalidad determinada, lo que puede resultar interesante para ciertas situaciones. Por ejemplo, si queremos saber su opinión sobre una película, podríamos preguntarle directamente.
    • Prompt: ”Dame tu opinión sobre la película interestelar”
    • Respuesta: “Como modelo de lenguaje de IA, no tengo emociones o opiniones personales, pero puedo darte información objetiva sobre la película Interestelar. Es una película de…”
Sin embargo, es importante tener en cuenta que como modelo de lenguaje de IA, ChatGPT no tiene emociones ni opiniones personales propias.
Para obtener una respuesta más específica, podemos pedirle que actúe como un experto en la materia.
    • Prompt: ”Quiero que actúes como un critico de cine. Desarrollarás una opinión atractiva y creativa de una película. Puedes abordar temas como la trama, el tono, la actuación y los personajes, la dirección, la música, los efectos de la película, la edición, el ritmo y el dialogo. Lo más importante es enfatizar cómo la película te ha hecho sentir. lo que realmente ha resonado contigo. También puedes ser crítico con la película. Por favor evita spoilers. Mi experiencia con “escribe tu opinión sobre la película interestelar”.”.
    • Respuesta: “La película Interestelar es un ejemplo impresionante de cómo el cine puede combinar una historia emocionalmente resonante con efectos visuales espectaculares y una dirección virtuosa. Desde el principio, la trama se establece como un viaje épico que atraviesa el tiempo y el espacio, y la película no decepciona en ningún aspecto.
La actuación de todo el reparto es…”
En este caso, ChatGPT responderá con una opinión más detallada y profunda, proporcionando una crítica más completa sobre la película. Esto es posible gracias a que podemos definir y personalizar cómo queremos que ChatGPT actúe y se comunique.
Esta herramienta es útil en caso de necesitar una opinión objetiva dentro de las capacidades de este, sobre cualquier tema, o incluso para interactuar de una forma más personalizada con ChatGPT en una variedad de situaciones, lo que nos permitirá aprovechar esto en una infinidad de casos interesantes o específicos, como pedirle ideas apara algo actuando como un experto en la materia y mucho más.

### **12.Creando contenido extenso**

1. Elección de prompts adecuados
El prompt es el texto que proporcionas a ChatGPT para que genere una respuesta. Es fundamental que sea claro y específico, ya que esto permite que el modelo entienda mejor el tema y entregue información relevante y precisa.

Ejemplo poco claro:
"Motor de automóvil"

Ejemplo claro y efectivo:
"¿Cómo funciona el motor de un automóvil?"

Cuanto más específico sea el prompt, mejor será la calidad y extensión de la respuesta.

2. Proporciona contexto adicional
ChatGPT genera respuestas basadas en un modelo preentrenado con gran cantidad de datos, pero no tiene conocimiento de contexto externo inmediato. Por eso, es importante que agregues información o detalles en el prompt para guiarlo y que produzca textos más precisos y relevantes.

Ejemplo sin contexto:
"Inteligencia artificial"

Ejemplo con contexto:
"¿Cuáles son las aplicaciones de la inteligencia artificial en la medicina?"

Al incluir detalles adicionales, el contenido será más profundo y enfocado.

3. Experimenta con hiperparámetros
Los hiperparámetros controlan cómo ChatGPT genera texto, ajustando aspectos como creatividad y precisión.

Temperatura: controla el grado de aleatoriedad. Valores bajos (ej. 0.2) generan respuestas más concretas y técnicas. Valores altos (ej. 0.8) producen respuestas más creativas y variadas.

Top-p (nucleus sampling): limita las opciones consideradas para la generación, equilibrando coherencia y diversidad.

Probar diferentes configuraciones te permite encontrar el equilibrio ideal para tu tipo de contenido, ya sea técnico, formal o creativo.

4. Revisión y corrección de errores
Aunque ChatGPT es avanzado, puede generar respuestas con errores o inexactitudes. Es importante revisar cuidadosamente el contenido generado, corregir posibles fallos y, si es posible, dar retroalimentación para mejorar resultados futuros.

Puedes pedir a ChatGPT que explique o amplíe alguna parte que no haya quedado clara.

Si detectas errores, corrígelos manualmente o solicita una nueva versión basada en tus observaciones.

5. Cómo crear contenido extenso paso a paso
Para lograr textos largos y bien estructurados, sigue esta estrategia:

Divide el tema en secciones: Introducción, desarrollo (con varios puntos o subtemas), y conclusión.

Solicita a ChatGPT que desarrolle cada sección con detalle.

Pide ejemplos, comparaciones o explicaciones adicionales para enriquecer el texto.

Itera pidiendo ampliaciones o reescrituras hasta alcanzar la extensión y calidad deseadas.

Utiliza listas, subtítulos y formato para organizar mejor el contenido.

# Herramientas de Análisis de Datos: Definición y Métricas Financieras Clave

## Introducción

En la era digital actual, el análisis de datos se ha convertido en una herramienta esencial para la toma de decisiones en las organizaciones. En el ámbito financiero, la correcta interpretación de grandes volúmenes de datos puede marcar la diferencia entre el éxito y el fracaso. Por ello, conocer las herramientas de análisis de datos y las métricas financieras clave es fundamental para profesionales y estudiantes que desean optimizar la gestión financiera y estratégica de una empresa.

## Definición de herramientas de análisis de datos

Las herramientas de análisis de datos son softwares o plataformas que permiten recopilar, procesar y analizar grandes cantidades de información para extraer insights valiosos. Estas herramientas facilitan la identificación de patrones, tendencias y anomalías que pueden influir en la toma de decisiones financieras. Algunos ejemplos incluyen Excel, Power BI, Tableau, y lenguajes de programación como Python y R.

## Principales herramientas utilizadas en análisis financiero

- **Microsoft Excel:** Muy utilizado por su versatilidad para organizar datos, realizar cálculos y generar gráficos.
- **Power BI y Tableau:** Plataformas de Business Intelligence que permiten crear dashboards interactivos y visualizaciones avanzadas.
- **Python y R:** Lenguajes de programación que ofrecen librerías especializadas para análisis estadístico, modelado financiero y machine learning.
- **SQL:** Lenguaje para gestionar bases de datos relacionales y extraer datos relevantes para el análisis.

## Métricas financieras clave

Algunas de las métricas más importantes que se analizan con estas herramientas son:

- **ROI (Return on Investment):** Mide la rentabilidad de una inversión.
- **EBITDA (Earnings Before Interest, Taxes, Depreciation and Amortization):** Indica la capacidad operativa de la empresa.
- **Margen de beneficio bruto:** Relación entre ingresos y costos directos.
- **Liquidez corriente:** Capacidad de la empresa para cumplir con sus obligaciones a corto plazo.
- **Rotación de inventarios:** Frecuencia con la que se renueva el inventario durante un periodo.

## Ejemplos prácticos de uso en la toma de decisiones financieras

- **Análisis de ventas:** Utilizando Power BI para identificar productos con mejor rendimiento y estacionalidad en las ventas.
- **Gestión de costos:** Aplicando Excel para controlar gastos y calcular márgenes de ganancia.
- **Predicciones financieras:** Empleando Python para crear modelos que pronostiquen ingresos futuros y posibles riesgos.
- **Optimización de inventarios:** Usando métricas de rotación para ajustar niveles de stock y reducir costos.

## Conclusión

Las herramientas de análisis de datos y las métricas financieras clave son pilares fundamentales para una gestión financiera efectiva. Su correcta aplicación permite a las empresas tomar decisiones informadas, anticiparse a cambios del mercado y maximizar sus recursos. Por ello, es imprescindible que los profesionales desarrollen competencias en estas áreas para afrontar los retos actuales del mundo financiero.

```markdown
Por favor, a partir de los siguientes datos financieros trimestrales:

| Trimestre | Ingresos | Costos | Activos Corrientes | Pasivos Corrientes | Utilidades Netas | Inversión Total |
|-----------|----------|--------|--------------------|--------------------|------------------|-----------------|
| Q1        | 500000   | 300000 | 120000             | 80000              | 70000            | 250000          |
| Q2        | 600000   | 350000 | 130000             | 90000              | 85000            | 260000          |
| Q3        | 550000   | 320000 | 125000             | 85000              | 75000            | 255000          |
| Q4        | 650000   | 360000 | 135000             | 95000              | 90000            | 265000          |

Calcula para cada trimestre los siguientes KPIs financieros:

- Margen bruto = (Ingresos - Costos) / Ingresos  
- ROI (Return on Investment) = Utilidades Netas / Inversión Total  
- Liquidez corriente = Activos Corrientes / Pasivos Corrientes  

Presenta los resultados en una tabla con los KPIs por trimestre y explica brevemente qué mide cada KPI.

# KPIs Financieros Calculados por Trimestre

A continuación se presentan los KPIs financieros calculados para cada trimestre, junto con una breve explicación de cada indicador.

| Trimestre | Margen Bruto (%) | ROI (%)        | Liquidez Corriente |
|-----------|------------------|----------------|--------------------|
| Q1        | 40.00            | 28.00          | 1.50               |
| Q2        | 41.67            | 32.69          | 1.44               |
| Q3        | 41.82            | 29.41          | 1.47               |
| Q4        | 44.62            | 33.96          | 1.42               |

---

## Explicación de los KPIs

- **Margen Bruto:**  
  Mide la rentabilidad de las ventas después de descontar los costos directos. Se calcula como:  
  \[
  \text{Margen Bruto} = \frac{\text{Ingresos} - \text{Costos}}{\text{Ingresos}} \times 100
  \]  
  Un margen bruto más alto indica mayor eficiencia en la producción o adquisición de bienes.

- **ROI (Return on Investment):**  
  Mide la rentabilidad de la inversión realizada en la empresa. Se calcula como:  
  \[
  ROI = \frac{\text{Utilidades Netas}}{\text{Inversión Total}} \times 100
  \]  
  Un ROI mayor indica que la inversión está generando buenos retornos.

- **Liquidez Corriente:**  
  Indica la capacidad de la empresa para cubrir sus obligaciones a corto plazo con sus activos corrientes. Se calcula como:  
  \[
  \text{Liquidez Corriente} = \frac{\text{Activos Corrientes}}{\text{Pasivos Corrientes}}
  \]  
  Un valor superior a 1 sugiere que la empresa puede cumplir con sus deudas inmediatas.

---

Si necesitas que te ayude a crear más KPIs o explicaciones, avísame.


# Clase 14: Formulación de prompts para el análisis de tendencias y visualización de KPIs

---

## Introducción

El análisis de tendencias y la visualización de KPIs (Indicadores Clave de Desempeño) son esenciales para comprender la evolución financiera o de negocio a lo largo del tiempo. Utilizar prompts bien diseñados en herramientas como ChatGPT permite generar insights valiosos y sugerencias para crear gráficos o reportes claros y efectivos.

En esta clase aprenderás cómo formular prompts efectivos para analizar tendencias en KPIs y para generar descripciones o instrucciones que faciliten su visualización.

---

## ¿Qué es el análisis de tendencias?

El análisis de tendencias consiste en examinar datos históricos para identificar patrones, comportamientos y movimientos que se repiten o evolucionan con el tiempo. Esto ayuda a prever futuros resultados y tomar decisiones informadas.

---

## Diseño de prompts para análisis de tendencias en KPIs

Para que ChatGPT te ayude a analizar tendencias, el prompt debe incluir:

- Los datos históricos o un resumen claro de los KPIs en diferentes períodos.
- Qué tipo de tendencia buscas (crecimiento, estacionalidad, ciclos, anomalías).
- Preguntas específicas para guiar el análisis (por ejemplo, ¿cuál es la tendencia general? ¿hay picos o caídas importantes?).

### Ejemplo de prompt para análisis de tendencias



## Buenas prácticas

- Siempre proporciona los datos claros y bien organizados.
- Define claramente el objetivo del análisis o visualización.
- Solicita explicaciones o interpretaciones para entender mejor los resultados.
- Si usas visualizaciones externas, indica qué herramientas o software usarás para adaptar las instrucciones.

---

## Ejemplo completo de prompt para ChatGPT

```markdown
Tengo los siguientes datos trimestrales de KPIs:

| Trimestre | Ventas Netas (miles USD) | Margen Bruto (%) | ROI (%) |
|-----------|--------------------------|------------------|---------|
| Q1 2022   | 200                      | 40               | 12      |
| Q2 2022   | 220                      | 42               | 15      |
| Q3 2022   | 210                      | 39               | 14      |
| Q4 2022   | 230                      | 44               | 16      |

1. Analiza la tendencia de cada KPI en el período indicado.  
2. Señala si hay crecimientos, caídas o fluctuaciones relevantes y posibles causas.  
3. Sugiere qué tipo de gráfico sería mejor para visualizar cada KPI y por qué.  
4. Describe paso a paso cómo crear el gráfico de líneas para la evolución del ROI trimestral, incluyendo título, etiquetas, y colores recomendados.

# Análisis de tendencias y visualización de KPIs financieros

## 1. Análisis de la tendencia de cada KPI

### Ventas Netas (miles USD)
Las ventas netas muestran una tendencia general de crecimiento a lo largo de los cuatro trimestres, pasando de 200 a 230 mil dólares. Aunque en el tercer trimestre hubo una leve caída a 210 mil, el incremento en el cuarto trimestre sugiere una recuperación y crecimiento sostenido. Esta fluctuación podría deberse a factores estacionales o campañas de marketing puntuales.

### Margen Bruto (%)
El margen bruto también presenta una tendencia positiva, aumentando de 40% en el primer trimestre a 44% en el último. La caída leve en el tercer trimestre (39%) coincide con la disminución en ventas netas, lo que puede indicar mayores costos o menor eficiencia temporal. En general, la empresa mejora su rentabilidad sobre ventas.

### ROI (%)
El ROI muestra un crecimiento constante, de 12% a 16%, reflejando que la rentabilidad de las inversiones está mejorando trimestre a trimestre. Esta tendencia positiva sugiere una gestión eficiente del capital invertido.

---

## 2. Fluctuaciones y posibles causas

- La caída en ventas netas y margen bruto en el tercer trimestre podría deberse a factores estacionales, problemas logísticos o competencia temporal.
- La recuperación en el cuarto trimestre indica acciones correctivas, mayor demanda o promociones exitosas.
- El crecimiento constante en ROI señala una mejor utilización de los recursos financieros.

---

## 3. Sugerencia de gráficos para visualización

- **Ventas Netas:** Gráfico de líneas para mostrar la evolución y tendencia a lo largo del tiempo.
- **Margen Bruto:** Gráfico de barras o líneas para destacar los cambios porcentuales entre trimestres.
- **ROI:** Gráfico de líneas para visualizar el crecimiento constante y facilitar comparaciones.

---

## 4. Instrucciones para crear un gráfico de líneas del ROI trimestral

1. **Selecciona la herramienta de visualización:** Puede ser Excel, Power BI, Tableau, Google Sheets u otro software de gráficos.
2. **Prepara los datos:** Organiza los trimestres en el eje X (horizontal) y los valores de ROI en el eje Y (vertical).
3. **Crea el gráfico de líneas:** Inserta un gráfico de líneas usando los datos organizados.
4. **Agrega título:** "Evolución del ROI Trimestral"
5. **Etiqueta los ejes:**  
   - Eje X: "Trimestre"  
   - Eje Y: "ROI (%)"
6. **Colores:** Usa un color azul para la línea principal, con marcadores visibles en cada punto de datos.
7. **Leyenda:** Si solo hay una serie, la leyenda puede omitirse para evitar distracciones.
8. **Opcional:** Añade líneas de tendencia o anotaciones para destacar crecimientos o eventos importantes.

---

¿Quieres que te ayude a crear los gráficos o preparar informes con estos datos?  


### **15.Evaluación de la efectividad de indicadores mediante prompts comparativos**

# Clase 15: Evaluación de la efectividad de indicadores mediante prompts comparativos

---

## Introducción

Los indicadores clave de desempeño (KPIs) son esenciales para medir el éxito y la salud financiera o operativa de una organización. Sin embargo, no todos los KPIs tienen la misma efectividad para diferentes contextos o metas empresariales. Por eso, es crucial **evaluar la efectividad de los indicadores** para asegurarnos de que realmente reflejan el desempeño y aportan valor a la toma de decisiones.

En esta clase aprenderemos cómo utilizar **prompts comparativos** en herramientas como ChatGPT para analizar y evaluar la efectividad de distintos KPIs, facilitando la identificación de los indicadores más útiles y relevantes.

---

## ¿Qué es un prompt comparativo?

Un prompt comparativo es una instrucción que solicita a ChatGPT o a otro modelo de lenguaje analizar dos o más elementos, resaltando diferencias, ventajas, desventajas o recomendaciones basadas en criterios específicos. 

En el contexto de KPIs, permite comparar indicadores para entender cuál es más efectivo en determinada situación.

---

## Cómo diseñar prompts comparativos para evaluar KPIs

Para evaluar la efectividad de indicadores usando prompts comparativos, es importante incluir:

- **Los KPIs a comparar**, con sus definiciones o fórmulas.
- **El contexto o área de aplicación** (ejemplo: ventas, finanzas, operaciones).
- **Criterios para evaluar efectividad**, como relevancia, facilidad de medición, impacto en decisiones, etc.
- **Una solicitud clara para identificar fortalezas y debilidades** de cada KPI y recomendar el más adecuado.

---

## Ejemplo de prompt comparativo para KPIs financieros

Compara los siguientes KPIs financieros usados para medir la rentabilidad de una empresa: Margen de Beneficio Neto y Retorno sobre Activos (ROA).

Para cada indicador, explica:

Qué mide exactamente.

Ventajas y limitaciones.

En qué contextos es más útil.

Cuál recomendarías para evaluar la rentabilidad general de una empresa mediana y por qué.


---

## Buenas prácticas

- Proporciona definiciones claras para cada KPI involucrado.
- Contextualiza la comparación para que el análisis sea relevante.
- Solicita ejemplos prácticos o escenarios para ilustrar diferencias.
- Pide una recomendación final o conclusión basada en el análisis.

---

## Ejemplo completo de prompt para ChatGPT

```markdown
Compara estos tres KPIs financieros: 

1. Margen Bruto  
2. Margen Operativo  
3. Margen Neto  

Para cada uno, describe:

- Qué mide y cómo se calcula.  
- Sus ventajas y limitaciones.  
- En qué casos o industrias es más útil.  

Finalmente, indica cuál es el más efectivo para analizar la rentabilidad en una empresa de retail y por qué.
