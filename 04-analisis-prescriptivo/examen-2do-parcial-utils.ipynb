```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Set random seed for reproducibility
np.random.seed(42)

# ------------------------------------
# Dataset 1: New Users Forecasting
# ------------------------------------
def generate_new_users_data():
    months = pd.date_range(start="2021-01-01", end="2023-12-31", freq="M")
    month_numbers = months.month  # Extract month numbers for seasonal effects

    base_users = 1000 + 50 * np.sin(2 * np.pi * np.arange(len(months)) / 12)

    winter_effect = np.where(month_numbers.isin([12, 1, 2]), 200, 0)  # Boost for winter months
    rainy_season_effect = np.where(month_numbers.isin([6, 7, 8]), 150, 0)  # Boost for rainy months

    seasonal_effect = winter_effect + rainy_season_effect

    trend = np.linspace(0, 600, len(months))  # Upward trend
    noise = np.random.normal(0, 50, len(months))  # Random noise

    users = base_users + seasonal_effect + trend + noise

    new_users_data = pd.DataFrame({'month': months, 'new_users': np.round(users)})
    new_users_data.to_csv("data/new_users_forecast.csv", index=False)

# ------------------------------------
# Dataset 2: Customer Retention (SEM)
# ------------------------------------
def generate_retention_data():
    n = 500
    # Engagement factors
    tiempo_promedio = np.random.normal(20, 5, n)
    interacciones = np.random.poisson(10, n)
    sesiones_activas = np.random.normal(15, 4, n)
    engagement = (0.4 * tiempo_promedio + 0.3 * interacciones + 0.3 * sesiones_activas)

    # Support factors
    tiempo_respuesta = np.random.normal(2, 0.5, n)  # lower is better
    resolucion_problemas = np.random.uniform(0, 1, n)  # higher is better
    soporte = (0.5 * resolucion_problemas - 0.5 * tiempo_respuesta)

    # Retention
    retencion = (0.6 * engagement + 0.4 * soporte + np.random.normal(0, 0.5, n))

    retention_data = pd.DataFrame({
        "tiempo_promedio": tiempo_promedio,
        "interacciones": interacciones,
        "sesiones_activas": sesiones_activas,
        "tiempo_respuesta": tiempo_respuesta,
        "resolucion_problemas": resolucion_problemas,
        "retencion": retencion
    })
    retention_data.to_csv("data/retention_data.csv", index=False)

# ------------------------------------
# Dataset 3: Historical ARPU and Conversion
# ------------------------------------
def generate_pricing_ab_test_data():
    # Historical
    n_arpu_hist = 1602
    arpu_hist = np.random.normal(2000, 300, n_arpu_hist)
    arpu_data_hist = pd.DataFrame({'customer_id': np.arange(n_arpu_hist), 'arpu': arpu_hist})
    arpu_data_hist.to_csv("data/correcaminos_historical_arpu.csv", index=False)
    
    n_conv_hist = 7857
    conversion_hist = np.random.binomial(1, 0.10, n_conv_hist)
    conversion_data_hist = pd.DataFrame({'lead_id': np.arange(n_conv_hist), 'converted': conversion_hist})
    conversion_data_hist.to_csv("data/correcaminos_historical_conversion.csv", index=False)
    
    # Test ARPU
    n_arpu_A = 980
    n_arpu_B = 560
    arpu_A = np.random.normal(2000, 300, n_arpu_A)
    arpu_B = np.random.normal(2250, 100, n_arpu_B)  # Increase due to pricing
    group = ['A'] * n_arpu_A + ['B'] * n_arpu_B
    arpu_data = pd.DataFrame({'group': group, 'arpu': np.concatenate([arpu_A, arpu_B])})
    arpu_data.to_csv("data/correcaminos_resultado_arpu_ab_test.csv", index=False)

    # Test Conversion
    n_conv_A = 10000
    n_conv_B = 7800
    conversion_A = np.random.binomial(1, 0.10, n_conv_A)
    conversion_B = np.random.binomial(1, 0.102, n_conv_B)
    group_conversion = ['A'] * n_conv_A + ['B'] * n_conv_B
    conversion_data = pd.DataFrame({'group': group_conversion, 'converted': np.concatenate([conversion_A, conversion_B])})
    conversion_data.to_csv("data/correcaminos_resultado_conversion_ab_test.csv", index=False)


# ------------------------------------
# Dataset 4: Marketing Channels (Regression and Optimization)
# ------------------------------------
def generate_marketing_data():
    # TikTok data
    tiktok_users = np.round(np.random.uniform(100,500,350))
    tiktok_costs = 10 + 0.05 * tiktok_users + np.random.normal(0, 5, len(tiktok_users))
    tiktok_data = pd.DataFrame({'channel': 'TikTok', 'users': tiktok_users, 'cost': tiktok_costs})

    # Meta data
    meta_users = np.round(np.random.uniform(100,500,655))
    meta_costs = 15 + 0.07 * meta_users + np.random.normal(0, 5, len(meta_users))
    meta_data = pd.DataFrame({'channel': 'Meta', 'users': meta_users, 'cost': meta_costs})

    # Combine and export
    marketing_data = pd.concat([tiktok_data, meta_data], ignore_index=True)
    marketing_data.to_csv("data/marketing_data.csv", index=False)

# Generate all datasets
generate_new_users_data()
generate_retention_data()
generate_pricing_ab_test_data()
generate_marketing_data()
```
