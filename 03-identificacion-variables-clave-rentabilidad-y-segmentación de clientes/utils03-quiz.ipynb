```python
import numpy as np
import pandas as pd
from datetime import datetime
import random

# Generar ecommerce_sales_data.csv (Datos de ventas con estacionalidad)
def generar_datos_ventas_con_estacionalidad():
    # Crear un rango de fechas mensuales desde hace 5 años hasta la fecha actual
    fechas = pd.date_range(end=datetime.today(), periods=5*12, freq='M')

    # Estacionalidad: Mayor ventas en noviembre y diciembre (temporada navideña)
    ventas_base = 5000 + np.arange(len(fechas)) * 10  # Tendencia creciente
    estacionalidad = np.sin(2 * np.pi * fechas.month / 12) * 1000  # Ciclo estacional anual
    ruido = np.random.normal(0, 300, size=len(fechas))  # Variabilidad aleatoria

    # Sumar estacionalidad y ruido a las ventas
    ventas = ventas_base + estacionalidad + ruido
    ventas = np.maximum(ventas, 3000)  # Asegurarse de que las ventas no bajen de un mínimo

    # Crear el DataFrame
    df_ventas = pd.DataFrame({'Fecha': fechas, 'Ventas': ventas})
    df_ventas.to_csv('data/ecommerce_sales_data.csv', index=False)
    return df_ventas

# Generar customer_satisfaction_data.csv con relaciones más coherentes
def generar_datos_satisfaccion_relaciones():
    np.random.seed(42)
    n_clientes = 500

    # Variables de satisfacción generadas con relaciones más coherentes
    tiempo_respuesta = np.random.normal(loc=7, scale=2, size=n_clientes)  # Escala de 1 a 10
    calidad_producto = np.random.normal(loc=8, scale=1.5, size=n_clientes)  # Escala de 1 a 10

    # Probabilidad de recomendación tiene una correlación negativa con el tiempo de respuesta
    prob_recomendacion = np.clip(10 - 0.5 * tiempo_respuesta + np.random.normal(0, 1, size=n_clientes), 1, 10)

    # LTV depende positivamente de calidad del producto y probabilidad de recomendación, y negativamente del tiempo de respuesta
    ltv = 100 * calidad_producto + 50 * prob_recomendacion - 20 * tiempo_respuesta + np.random.normal(0, 10, n_clientes)

    # Crear el DataFrame
    df_satisfaccion = pd.DataFrame({
        'Tiempo_Respuesta': np.clip(tiempo_respuesta, 1, 10),  # Limitar a escala 1-10
        'Calidad_Producto': np.clip(calidad_producto, 1, 10),  # Limitar a escala 1-10
        'Probabilidad_Recomendacion': np.clip(prob_recomendacion, 1, 10),  # Limitar a escala 1-10
        'LTV': np.round(ltv, 2)  # Valor total esperado de vida del cliente
    })
    df_satisfaccion.to_csv('data/customer_satisfaction_data.csv', index=False)
    return df_satisfaccion

# Generar los archivos de datos
ventas_data = generar_datos_ventas_con_estacionalidad()
satisfaccion_data = generar_datos_satisfaccion_relaciones()

print("Archivos ecommerce_sales_data.csv y customer_satisfaction_data.csv generados exitosamente.")
```
