#### `OBJETIVO GENERAL `
>  <p style='text-align: justify;'> Este laboratorio tiene el fin de analizar datos financieros para definir métricas clave, evaluar la madurez analítica de la organización, identificar variables críticas que impulsan la rentabilidad e incorporar modelos matemáticos, con el fin de mejorar la toma de decisiones en finanzas y negocios.</p>

`Módulo 1.` **Herramientas de análisis de datos y definición de métricas financieras clave.**

> <p style='text-align: justify;'> Objetivo: Emplear herramientas de análisis de datos para establecer métricas financieras clave, garantizando su pertinencia y claridad en la evaluación del desempeño organizacional.</p>

1. ¿Qué es un indicador? ¿Que son los KPIs, Balanced Scorecard y OKRs? ¿Cómo se clasifican?
    - Métodos del caso: 
        - Lenguaje de trabajo (Python y pandas).
        - Definición de un KPI relevante y calculo.
    - Reporte de laboratorio 1.
2. Python para cálculo de indicadores.
    - Métodos del caso: 
        - Contestar preguntas de negocio usando Python.
        - Cálculo de indicadores en entorno profesional.
3. Python para limpieza de datos.
    - Métodos del caso: 
        - Limpieza de nulos y duplicados.
        - Tratamiento de outliers.
4. Python y AI para análisis de datos (Quiz 1).
    - Métodos del caso: 
        - Realizar análisis usando ChatGPT Canvas.
        - Uso de AI para análisis de datos.
    - Reporte de laboratorio 2.

`Módulo 2.`  **Evaluación de la madurez analítica y elaboración de reportes ejecutivos.**

> <p style='text-align: justify;'> Objtivo: Evaluar la madurez analítica de la empresa y elaborar reportes de valor que orienten la toma de decisiones basadas en evidencia.</p>

1. Análisis de madurez analítica de una empresa y tipos de análisis.
    - Método del caso: 
        - Selección de KPI de acuerdo a madurez de la empresa.
        - Recomendación de análisis de acuerdo a objetivo de la empresa.
2. Elaboración de reportes y presentaciones ejecutivas
    - Método del caso:
        - Presentación de resultados de análisis.
    - Reporte de laboratorio 3.
3. Estadísticos, distribución, covarianza y correlación en entorno profesional. 
    - Método del caso: 
        - Comprender los patrones de indicadores.
        - Medición de relación entre indicadores.
4. Técnicas de visualización de datos con Matplotlib y Seaborn (Quiz 2). 
    - Método del caso: 
        - Visualizaciones para la toma de decisiones.
    - Reporte de laboratorio 4.
    - Quiz 2
    - Clase de repaso.
    - Examen de medio término.

`Módulo 3.`  **Identificación de variables clave de rentabilidad y segmentación de clientes.**

> <p style='text-align: justify;'> Objetivo: Identificar variables determinantes de la rentabilidad y segmentar a los clientes para optimizar recursos y maximizar el desempeño empresarial.</p>

1. Analítica multivariable.
    - Método del caso: 
        - Segmentos de clientes en la empresa.
        - Uso de embedding vectors para análisis de texto.
2. Regresiones multivariables sobre indicadores en un entorno profesional.
    - Método del caso: 
        - Descubrir variables que impactan a KPI.
        - Predecir LTV de un cliente.
    - Reporte de laboratorio 5.
3. Definición de análisis predictivo y series de tiempo (Prophet, Autoarima). 
    - Método del caso: 
        - Predicción de google trends para share of search.
        - Predicción de ventas en una empresa.
4. Uso de modelos predictivos para toma de decisiones.
    - Método del caso: 
        - De acuerdo a un modelo predictivo de negocio plasmar los posibles resultados a observar.
        - Optimización de presupuesto para marketing.
    - Reporte de laboratorio 6.
    - Quiz 3.

`Módulo 4.`  **Diseño de indicadores accionables con modelos matemáticos y mejora continua.**

> <p style='text-align: justify;'> Diseñar indicadores a partir de modelos matemáticos que permitan generar recomendaciones accionables, impulsando la mejora continua en la toma de decisiones organizacionales.</p>

1. Variables latentes y sistema de ecuaciones estructurales (SEM) en el contexto profesional (semopy).
    - Método del caso: 
        - Encontrar al cliente mas leal.
        - Dar una conclusión de negocio usando variables latentes y SEM
    - Reporte de laboratorio 7.
    - Quiz siguiente clase.
2. Pruebas A/B en entorno profesional.
    - Método del caso: 
        - Generar un plan de acción para modificar el precio.
        - Análisis de resultados de AB testing
    - Reporte de laboratorio 8.
3. Evaluación de conocimientos
    - Clase de repaso
    - Examen final
    - Presentación de proyecto
