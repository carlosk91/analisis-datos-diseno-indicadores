**2. Características de una buena presentación ejecutiva:**  

- **Claridad**: Ser directo y evitar información innecesaria.<br>
- **Concisión**: Resumir lo esencial para poder tomar una decisión (dependiendo del tema y el objetivo de la presentación).<br>
- **Visualización:** Usar gráficos y datos para hacer la información comprensible de forma rápida.<br>
- **Enfoque en los resultados y las recomendaciones:** La audiencia debe salir con una idea clara de lo que se espera que hagan después de la presentación.

### **3. Cómo Crear una Presentación Ejecutiva (10 minutos)**  

#### **Estructura básica de la presentación:**

- **Introducción**: Contexto y objetivos de la presentación.
- **Desarrollo**: Principales hallazgos, análisis, y visualización de datos.
- **Conclusión**: Resumen, recomendaciones, y próximos pasos.

#### **La importancia del storytelling en una presentación ejecutiva:**

##### **¿Qué es el storytelling?**
El storytelling es la técnica de contar una historia para conectar emocionalmente con tu audiencia. En presentaciones ejecutivas, ayuda a transmitir datos complejos de forma que se comprendan y sean impactantes.

<center>
 <img src="https://www.blastanalytics.com/wp-content/uploads/2020/10/data-storytelling-training-elements.png"  width="300" />
</center> 

##### **Cómo integrar storytelling:**
1. **Comienza con una narrativa clara**: Introduce un reto o situación que necesita una solución (diagnóstico).
2. **Presenta los datos como personajes**: Los datos deben ser presentados como elementos que respaldan la solución propuesta, no como números aislados.
3. **Cierra con una llamada a la acción**: Termina con una conclusión clara y concreta que resuma las decisiones que la audiencia debe tomar.

 **Diseño visual en presentaciones ejecutivas:**
- Uso de colores, tipografías y gráficas para reforzar el mensaje.
- Eliminar elementos visuales innecesarios y centrarse en lo que realmente importa.

### **5. Diseño de presentaciones - Tips and Tricks (15 minutos)**  

**UTILIZALO EN TU PROYECTO FINAL**

### **Distribución ideal de una presentación**

#### Las 5 reglas de las presentaciones efectivas

1. **Zonas de atención visual:**  
   - **Zona superior izquierda:** Es donde los espectadores suelen mirar primero (lectura en “Z” o “F”). Ideal para poner el título o el punto principal.  
   - **Zona superior derecha:** Segunda área de atención, buena para información secundaria o subtítulos.  
   - **Zona inferior izquierda y derecha:** Se miran después, aquí puedes poner gráficos, imágenes o detalles de apoyo.

2. **Regla del tercio (Regla de los tercios):**  
   Divide la diapositiva en 3 partes horizontales y 3 verticales, formando 9 cuadros. Los puntos donde se cruzan las líneas son ideales para colocar elementos importantes (títulos, imágenes clave, datos principales). Esto ayuda a equilibrar el diseño.

3. **Jerarquía visual:**  
   - Usa tamaños y colores para destacar lo más importante.  
   - Pocos puntos clave, evita saturar la slide.  
   - Texto claro, conciso y legible (fuente grande y buen contraste).

4. **Divide la información en secciones:**  
   - **Título o encabezado:** Arriba y claro.  
   - **Contenido principal:** Centro o lado izquierdo para que sea lo primero que se lea.  
   - **Soporte visual o datos:** Lado derecho o parte inferior.  
   - **Notas o referencias:** En la parte más baja, pequeño y discreto.

5. **Ley de la primacía y recencia:**  
   La audiencia recuerda mejor lo primero y lo último que ve, así que estructura tu presentación para aprovechar esto (introducción clara y conclusión fuerte).

#### **Gráficas recomendadas para presentaciones ejecutivas**

- Gráficas de barras
  - **Uso:** Comparar categorías o mostrar cambios a lo largo del tiempo.
  - **Utilidad:** Son simples, eficaces y facilitan la comparación.
  - **Ejemplo:** Comparar ventas de productos o rendimiento de sucursales.

- Gráficas de líneas
  - **Uso:** Mostrar la evolución de un dato a lo largo del tiempo.
  - **Utilidad:** Ideales para visualizar tendencias.
  - **Ejemplo:** Evolución de ventas mes a mes.

- Gráficas circulares (de pastel)
  - **Uso:** Mostrar la distribución de un todo entre varias categorías.
  - **Utilidad:** Representan cómo se dividen las partes de un todo.
  - **Ejemplo:** Participación de mercado de competidores.

- Gráficas de área
  - **Uso:** Mostrar el cambio de un valor a lo largo del tiempo, destacando magnitudes.
  - **Utilidad:** Visualiza la suma de varias series.
  - **Ejemplo:** Comparación de ingresos por línea de productos.

- Tablas con resúmenes visuales
  - **Uso:** Resumir datos y facilitar la comparación visual rápida.
  - **Utilidad:** Mejoran la legibilidad con colores o íconos.
  - **Ejemplo:** Datos financieros trimestrales con indicadores visuales.

- Gráficas de dispersión
  - **Uso:** Mostrar la relación entre dos variables.
  - **Utilidad:** Ayuda a observar correlaciones y tendencias.
  - **Ejemplo:** Relación entre inversión en marketing y aumento de ventas.

#### **Gráficas a evitar**

- Gráficas de barras 3D
  - **Por qué evitarlas:** Son difíciles de leer y pueden distorsionar la percepción.
  - **Alternativa:** Usa barras 2D simples.

- Gráficas circulares con demasiadas categorías
  - **Por qué evitarlas:** Son confusas con demasiados segmentos.
  - **Alternativa:** Usa barras o tablas.

- Gráficas de radar
  - **Por qué evitarlas:** Son complicadas de leer y difíciles de interpretar.
  - **Alternativa:** Usa tablas comparativas o barras.

- Gráficas de líneas con demasiadas series
  - **Por qué evitarlas:** Se sobrecargan y dificultan la lectura.
  - **Alternativa:** Limita las series a 3-4.

- Gráficas de dispersión sin etiquetas o puntos pequeños
  - **Por qué evitarlas:** Son ilegibles sin etiquetas o puntos claros.
  - **Alternativa:** Asegúrate de etiquetar y ajustar el tamaño de los puntos.

- Gráficas con escalas no uniformes
  - **Por qué evitarlas:** Pueden dar una percepción incorrecta de los datos.
  - **Alternativa:** Usa escalas uniformes y claras.

#### **Uso de Colores**

- **Paleta de Colores Coherente:**
  - **Limita la paleta de colores:** No utilices más de 3-4 colores principales en toda la presentación. Demasiados colores pueden generar confusión y distraer la atención de la audiencia.
  - **Usa los colores de la marca:** Si la presentación es para una empresa o institución, utiliza los colores corporativos para mantener la consistencia visual. Esto también fortalece la identidad de la marca.
  - **Colores neutrales para el fondo:** Colores como el blanco, gris claro o tonos suaves de azul son ideales para el fondo. Evita colores brillantes o saturados que puedan hacer que el texto sea difícil de leer.

- **Contrastes:**
  - **Contraste alto para la legibilidad:** Asegúrate de que el contraste entre el fondo y el texto sea suficiente. Por ejemplo, texto oscuro sobre fondo claro o texto blanco sobre fondo oscuro.
  - **Evita colores similares para texto y fondo:** No uses un fondo azul con texto azul claro, ya que esto dificultará la lectura.

- **Psicología del Color:**
  - **Azul:** Representa profesionalismo, confianza y tranquilidad. Es ideal para negocios y presentaciones ejecutivas.
  - **Rojo:** Llama la atención y puede usarse para destacar puntos importantes o alertas. Úsalo con moderación, ya que es un color intenso.
  - **Verde:** Asociado con el crecimiento, la calma y el éxito. Puede usarse para resaltar datos positivos.
  - **Amarillo:** Energético, pero debe usarse con cautela, ya que puede ser demasiado brillante y cansar a la vista si se usa en exceso.
  - **Negro o gris oscuro:** Son excelentes para texto y gráficos, proporcionando un aspecto limpio y serio.
