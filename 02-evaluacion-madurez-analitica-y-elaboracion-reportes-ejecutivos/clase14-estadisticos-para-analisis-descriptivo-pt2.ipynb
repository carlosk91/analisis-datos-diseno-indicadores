## Contenido

### 1. Introducción a la Correlación (15 minutos)
La correlación mide:
- **Fuerza:** Qué tan estrechamente están relacionadas dos variables.
- **Dirección:** Si una variable aumenta cuando la otra aumenta (correlación positiva) o disminuye (correlación negativa).

#### Ejemplo:
Supongamos que analizamos el gasto en promociones y las ventas en una empresa:
- **Correlación positiva:** A mayor gasto en promociones, mayores ventas.
- **Correlación negativa:** A mayor gasto en promociones, menores ventas (ineficiencia o saturación).

#### Coeficiente de correlación (Pearson):
- **1.0:** Correlación perfecta positiva.
- **0.0:** Sin correlación.
- **-1.0:** Correlación perfecta negativa.

---

### 2. Cómo Calcular y Visualizar Correlaciones (20 minutos)
Podemos usar una **matriz de correlación** para identificar relaciones significativas entre muchas variables.

#### Ejemplo en Python:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar dataset
data = pd.read_csv("data/modaexpress_ventas.csv")

# Calcular la matriz de correlación
correlation_matrix = data.corr(numeric_only=True)

# Visualizar la matriz
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de Correlación")
plt.show()
```

- Identifica patrones inesperados y posibles sesgos en los datos.

Analicemos un scatterplot para ver el impacto de las promociones en las ventas por región

```python
sns.scatterplot(x=data["Gasto en Promociones (USD)"], y=data["Ventas (USD)"], hue=data["Región"])
```

Ahora, veamos cómo impactan los descuentos en las ventas por categoría de producto

```python
plt.figure(figsize=(8, 4))
sns.barplot(x='Categoría de Producto', y='Ventas (USD)', hue='Descuento (%)', data=data)
plt.title('Comparación de Ventas con y sin Descuento por Categoría')
plt.xticks(rotation=45)
plt.legend(title='Descuento', labels=data['Descuento (%)'].unique(), loc="upper right")
plt.tight_layout()
plt.show()
```

### 4. Análisis Multivariable y Segmentación (30 minutos)
El análisis multivariable permite entender relaciones complejas considerando varios factores simultáneamente.

#### Ejemplo:
- La correlación entre las ventas y las promociones podría variar dependiendo de la región o la categoría de producto.

#### Visualización por Segmentos:

```python
# Relación entre gasto en promociones y ventas, segmentada por región
sns.lmplot(x="Gasto en Promociones (USD)", y="Ventas (USD)", hue="Región", data=data, scatter_kws={'alpha':0.3, 's': 5})
plt.title("Relación Promociones vs Ventas por Región")
plt.show()
```

```python
# Relación entre gasto en promociones y ventas, segmentada por categoría de producto
sns.lmplot(x="Gasto en Promociones (USD)", y="Ventas (USD)", hue="Categoría de Producto", data=data, scatter_kws={'alpha':0.3, 's': 5})
plt.title("Relación Promociones vs Ventas por Categoría de Producto")
plt.show()
```
