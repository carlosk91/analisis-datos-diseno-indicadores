# Sesión 9: Madurez Analítica Nula, Descriptiva y Diagnóstica

## **Objetivo de la sesión:**

Proporcionar a los estudiantes una comprensión profunda de las etapas de madurez analítica de una organización. Incluirán análisis de un ejercicio práctico para proponer mejoras basadas en las mejores prácticas.

---

### **1. Introducción al método del caso** (10 mins)

El Departamento de Estudios a Distancia de la Universidad Autónoma de Manizales (SEAD UAM) enfrenta serios desafíos en la gestión de su información y en la toma de decisiones basada en datos. Actualmente, la organización opera con sistemas fragmentados, lo que genera inconsistencias, errores y decisiones desinformadas. Además, la falta de automatización en los procesos y la ausencia de una cultura de datos limitan su capacidad para responder de manera efectiva a las necesidades de estudiantes y colaboradores.

Como parte del equipo de datos del SEAD UAM, tu responsabilidad será analizar esta situación y diseñar soluciones prácticas que permitan a la organización avanzar en su nivel de madurez analítica. **¿Qué herramientas y procesos implementarías para resolver los problemas actuales? ¿Cómo fomentarías una cultura de datos en un entorno tradicional?**

### **2. Madurez Analítica Baja y Conceptos Clave** (15 mins)

![Introducción a los niveles de madurez analítica](https://www.delalamo.es/wp-content/uploads/2020/06/From-Hindsight-to-Foresight-gartner-300x222-2.png)

#### **2.1 Niveles Bajos de Madurez Analítica**

1. **Nivel 0: Madurez Analítica Nula**

   - **Características:** Decisiones basadas en intuición o experiencia sin respaldo de datos. Falta de procesos estructurados para recopilar y analizar datos.
   - **Ejemplos:** Empresas familiares pequeñas que toman decisiones según prácticas tradicionales.
   - **Herramientas:** Hojas de cálculo locales como Excel.
   - **Uso:** Registro manual de datos y generación de informes básicos.
   - **Roles clave:** Asistentes administrativos encargados de recopilar y consolidar datos manualmente.
   - **Acciones recomendadas:**
     - Implementar sistemas básicos de recopilación de datos (como hojas de cálculo compartidas).
     - Capacitar en conceptos básicos de datos, como la organización y limpieza de información.

2. **Nivel 1: Analítica Descriptiva**

   - **Características:** Uso de datos para reportar qué ocurrió en el pasado. Generación de reportes estáticos sin capacidad de exploración.
   - **Ejemplo:** Un minorista que elabora informes trimestrales manuales de ingresos y gastos sin segmentación por clientes o productos.
   - **Herramientas:** Hojas de cálculo en nube (Google Sheets) y sistemas básicos de bases de datos.
   - **Uso:** Creación de reportes estáticos para monitorear métricas clave.
   - **Roles clave:** Analistas básicos responsables de generar reportes descriptivos y realizar análisis simples.
   - **Acciones recomendadas:**
     - Automatizar reportes mediante dashboards interactivos con herramientas como Power BI o Tableau.
     - Introducir formación en conceptos de visualización de datos y BI.

3. **Nivel 2: Analítica Diagnóstica**

   - **Características:** Análisis del porqué de los eventos ocurridos mediante correlaciones o identificación de patrones.
   - **Ejemplo:** Una compañía de software analiza la deserción de usuarios identificando patrones en datos de comportamiento y encuestas de satisfacción.
   - **Herramientas:** Plataformas de Business Intelligence como Power BI, Tableau o Metabase.
   - **Uso:** Análisis diagnóstico; visualización avanzada de datos para entender patrones y tendencias.
   - **Roles clave:** Analistas de datos y ingenieros de datos encargados de consolidar y analizar información de diversas fuentes.
   - **Acciones recomendadas:**
     - Implementar herramientas como Python o R para realizar análisis estadísticos.
     - Capacitar a empleados en pruebas de hipótesis y análisis causal.

#### **2.2 Conceptos Clave**

##### **1. Centralización de Datos: Problemas y Soluciones**

- **Definición:** La centralización de datos implica integrar toda la información de una organización en un único sistema accesible para todos los departamentos relevantes.
- **Problemas comunes con los silos de datos:**
  - Cada departamento almacena información en sistemas separados sin comunicación entre ellos.
  - **Ejemplo:** En una empresa de retail, el área de inventarios utiliza un sistema interno de gestión (ERP) para rastrear existencias, mientras que el equipo de marketing emplea una plataforma de CRM para planificar promociones. Estos sistemas no están conectados, lo que provoca discrepancias al coordinar campañas, ya que el área de marketing no tiene acceso en tiempo real a los niveles de inventario. Esto genera sobrepromociones de productos agotados o subutilización de productos disponibles, afectando las ventas y la experiencia del cliente.
- **Beneficios de la centralización:**
  - Información consistente y actualizada.
  - Colaboración más efectiva entre áreas.
  - Reducción de errores en la toma de decisiones.
- **Visualización:**
  - **Silos de datos:** Sistemas aislados que no comparten información.
  - **Datos centralizados:** Información consolidada en una base de datos relacional o un sistema de inteligencia de negocios.

##### **2. Automatización de Procesos: Beneficios y Ejemplos**

- **Definición:** La automatización reemplaza tareas manuales repetitivas con herramientas que recolectan y procesan datos automáticamente.
- **Ejemplo:** En lugar de consolidar datos manualmente en hojas de cálculo, una empresa utiliza Power BI para conectar su base de datos y generar dashboards automatizados.
- **Beneficios:**
  - Ahorro de tiempo.
  - Reducción de errores humanos.
  - Capacidad de generar reportes en tiempo real.

##### **3. Cultura de Datos: Creando un Entorno Basado en Evidencia**

- **Definición:** Es la práctica de tomar decisiones basadas en evidencia y análisis de datos, en lugar de intuiciones.
- **Cómo fomentarla:**
  - Capacitación en interpretación de datos y uso de herramientas analíticas.
  - Promoción del uso de KPIs.
  - Implementación de sistemas accesibles que permitan explorar y analizar datos fácilmente.
- **Ejemplo:** Una empresa de tecnología utiliza capacitaciones mensuales para que su equipo interprete métricas clave de desempeño, como el churn o la retención de usuarios.

Con estos recursos y conceptos, podrás analizar y proponer soluciones efectivas para el caso del SEAD UAM.

### **4. Madurez Analítica Alta y Conceptos Clave** (15 mins)

#### **Nivel 3: Analítica Predictiva**

- **Características:** Uso de modelos estadísticos y machine learning para predecir eventos futuros con base en datos históricos.
- **Ejemplo:** Un banco utiliza modelos predictivos para calcular la probabilidad de incumplimiento de pago en créditos hipotecarios.
- **Herramientas:** Lenguajes de programación para modelado estadístico y plataformas de machine learning. *¿Cuales lenguajes de programación se te ocurre que usan en las empresas?*
- **Roles clave:** Científicos de datos y analistas avanzados. Estos roles son responsables de construir y evaluar modelos predictivos aplicados a necesidades específicas.
- **Acciones recomendadas:**
  - Formar equipos interdisciplinarios que combinen habilidades técnicas y conocimiento del negocio.
  - Diseñar flujos de trabajo iterativos para evaluar modelos predictivos en escenarios controlados.

#### **Nivel 4: Analítica Prescriptiva**

- **Características:** Generación de recomendaciones para optimizar resultados. Utilización de algoritmos avanzados para automatizar la toma de decisiones.
- **Ejemplo:** Una empresa de transporte ajusta rutas y precios dinámicamente con base en demanda en tiempo real y costos operativos.
- **Herramientas:** Plataformas de optimización y sistemas de inteligencia artificial integrados.
- **Roles clave:** Ingenieros de machine learning y arquitectos de datos. Estos roles lideran la integración de sistemas avanzados en operaciones críticas.
- **Acciones recomendadas:**
  - Implementar sistemas de simulación que permitan experimentar con escenarios antes de tomar decisiones operativas.
  - Establecer mecanismos de retroalimentación para ajustar las recomendaciones automáticamente.

#### **Conceptos Clave**

##### **1. Modelos Predictivos**

- **Definición:** Algoritmos matemáticos y estadísticos que analizan datos históricos para prever futuros eventos o comportamientos.
- **Ejemplo:** Un modelo de churn predictivo identifica clientes con alta probabilidad de cancelar sus servicios.
- **Herramientas comunes:** Lenguajes y plataformas utilizadas para aprendizaje automático.

##### **2. Variables Predictoras y Target**

- **Definición:**
  - **Variables predictoras:** Factores independientes que ayudan a prever el resultado (por ejemplo, frecuencia de compra).
  - **Variable target:** Resultado que se desea predecir (por ejemplo, deserción de clientes).
- **Importancia:** Seleccionar las variables más relevantes mejora significativamente la precisión del modelo.

##### **3. Validación y Métricas de Evaluación**

- **Definición:** Métodos para medir el rendimiento de un modelo predictivo.
- **Métricas comunes:**
  - Precisión (Accuracy).
  - Curva ROC-AUC.
  - Error cuadrático medio (MSE).

##### **4. Algoritmos Prescriptivos**

- **Definición:** Técnicas avanzadas que no solo predicen eventos futuros, sino que recomiendan acciones óptimas.
- **Ejemplo:** Algoritmos que optimizan precios dinámicamente según la demanda y el inventario.
- **Herramientas comunes:** Plataformas de optimización de recursos y simulación avanzada.

##### **5. Simulaciones y Análisis de Escenarios**

- **Definición:** Métodos que evalúan múltiples escenarios para identificar la mejor estrategia.
- **Ejemplo:** Simular rutas logísticas para minimizar costos de transporte.

##### **6. Sistemas de Toma de Decisiones Automatizados**

- **Definición:** Plataformas integradas que ejecutan acciones basadas en los resultados de modelos prescriptivos.
- **Ejemplo:** Ajuste automático de precios en una plataforma de e-commerce.
- **Requisitos:**
  - Conexión en tiempo real a fuentes de datos.
  - Integración con sistemas operativos clave.

### **6. Caso de Estudio: Optimización en Retail Plus** (15 minutos)

### Caso de Estudio: Netflix y la Evolución hacia Recomendaciones Basadas en Datos

#### Escenario:

A principios de la década de 2000, Netflix enfrentaba un desafío significativo: mejorar su sistema de recomendaciones para ofrecer una experiencia personalizada y mantener la competitividad en el mercado de entretenimiento digital. Las recomendaciones genéricas, basadas únicamente en la popularidad de los títulos, no lograban captar los intereses individuales de los usuarios, lo que resultaba en insatisfacción, deserción y un bajo tiempo promedio de visualización.

En este contexto, se te asigna el rol de un especialista en datos dentro del equipo de Netflix, con el objetivo de diseñar soluciones que transformen el sistema de recomendaciones. Tu responsabilidad es diagnosticar las limitaciones actuales, identificar oportunidades en los datos disponibles y proponer un plan estratégico para implementar analítica predictiva que anticipe los intereses de los usuarios.

---

**Problemas identificados:**

- Falta de personalización en las recomendaciones ofrecidas a los usuarios.
- Los patrones históricos de visualización no eran analizados para generar insights significativos.
- La falta de un modelo predictivo impedía anticipar qué contenido sería relevante para cada usuario.

**Datos disponibles:**

- Historial de visualización de los usuarios.
- Datos básicos que reflejan preferencias generales y patrones de consumo.
- Calificaciones otorgadas por los usuarios a películas y series.
- Metadatos sobre el contenido, incluyendo género, elenco y director.

**Pregunta clave:** ¿Qué datos adicionales podrían ser cruciales para mejorar las recomendaciones y cómo se podrían recolectar?

**Objetivo:**
¿Cómo podría Netflix transformar su sistema de recomendaciones en una herramienta predictiva que anticipe los intereses individuales de cada usuario, mejore la satisfacción del cliente y aumente el tiempo de interacción con la plataforma?

---
