# Sesión 10: KPIs, Herramientas y Estrategias para Mejorar la Madurez Analítica

## **Objetivo de la sesión:**
Proporcionar a los estudiantes una comprensión profunda de herramientas y estrategias comprobadas para avanzar en la madurez analítica de una organización. Incluirán análisis de un ejercicio práctico para proponer mejoras basadas en las mejores prácticas.

---

### **1. Introducción al método del caso** (15 mins)

El Departamento de Estudios a Distancia de la Universidad Autónoma de Manizales (SEAD UAM) enfrenta serios desafíos en la gestión de su información y en la toma de decisiones basada en datos. Actualmente, la organización opera con sistemas fragmentados, lo que genera inconsistencias, errores y decisiones desinformadas. Además, la falta de automatización en los procesos y la ausencia de una cultura de datos limitan su capacidad para responder de manera efectiva a las necesidades de estudiantes y colaboradores.

Como parte del equipo de datos del SEAD UAM, tu responsabilidad será analizar esta situación y diseñar soluciones prácticas que permitan a la organización avanzar en su nivel de madurez analítica. **¿Qué herramientas y procesos implementarías para resolver los problemas actuales? ¿Cómo fomentarías una cultura de datos en un entorno tradicional?**

---

### **2. Conceptos y recursos clave para resolver el caso** (30 mins)

Esta sección presenta los conceptos, herramientas y procesos necesarios para diseñar soluciones efectivas. Aunque los ejemplos no están relacionados directamente con el SEAD UAM, los principios pueden aplicarse para resolver el caso.

#### **1. Centralización de datos**

- **Definición:** La centralización de datos implica integrar toda la información de una organización en un único sistema accesible para todos los departamentos relevantes.
- **Problemas comunes con los silos de datos:**
  - Cada departamento almacena información en sistemas separados sin comunicación entre ellos.
  - **Ejemplo:** En una empresa de retail, el área de inventarios utiliza un sistema interno de gestión (ERP) para rastrear existencias, mientras que el equipo de marketing emplea una plataforma de CRM para planificar promociones. Estos sistemas no están conectados, lo que provoca discrepancias al coordinar campañas, ya que el área de marketing no tiene acceso en tiempo real a los niveles de inventario. Esto genera sobrepromociones de productos agotados o subutilización de productos disponibles, afectando las ventas y la experiencia del cliente.
- **Beneficios de la centralización:**
  - Información consistente y actualizada.
  - Colaboración más efectiva entre áreas.
  - Reducción de errores en la toma de decisiones.
- **Visualización:**
  - **Silos de datos:** Sistemas aislados que no comparten información.
  - **Datos centralizados:** Información consolidada en una base de datos relacional o un sistema de inteligencia de negocios.

#### **2. Automatización de procesos**

- **Definición:** La automatización reemplaza tareas manuales repetitivas con herramientas que recolectan y procesan datos automáticamente.
- **Ejemplo:** En lugar de consolidar datos manualmente en hojas de cálculo, una empresa utiliza Power BI para conectar su base de datos y generar dashboards automatizados.
- **Beneficios:**
  - Ahorro de tiempo.
  - Reducción de errores humanos.
  - Capacidad de generar reportes en tiempo real.

#### **3. Cultura de datos**

- **Definición:** Es la práctica de tomar decisiones basadas en evidencia y análisis de datos, en lugar de intuiciones.
- **Cómo fomentarla:**
  - Capacitación en interpretación de datos y uso de herramientas analíticas.
  - Promoción del uso de KPIs.
  - Implementación de sistemas accesibles que permitan explorar y analizar datos fácilmente.
- **Ejemplo:** Una empresa de tecnología utiliza capacitaciones mensuales para que su equipo interprete métricas clave de desempeño, como el churn o la retención de usuarios.

#### **4. Herramientas analíticas y roles clave** (Roles incluidos)

- **Nivel 0-1:**

  - **Herramientas:** Hojas de cálculo como Excel o Google Sheets.
  - **Uso:** Generación de informes básicos y análisis descriptivos.
  - **Roles clave:** Analistas básicos o asistentes administrativos responsables de tareas manuales y reportes.

- **Nivel 2-3:**

  - **Herramientas:** Plataformas de Business Intelligence como Power BI o Tableau; lenguajes de programación como Python y R.
  - **Uso:** Análisis diagnóstico y predictivo; visualización avanzada de datos para entender patrones y tendencias.
  - **Roles clave:** Analistas de datos, ingenieros de datos y científicos de datos encargados de consolidar y analizar información de diversas fuentes.

- **Nivel 4:**

  - **Herramientas:** Soluciones avanzadas de inteligencia artificial y aprendizaje automático, como AWS SageMaker o Google AI.
  - **Uso:** Análisis prescriptivo; modelos predictivos complejos; automatización de decisiones en tiempo real.
  - **Roles clave:** Científicos de datos avanzados, ingenieros de machine learning y arquitectos de datos, que desarrollan y despliegan sistemas integrados de IA para soportar decisiones estratégicas.

#### **5. Diseño de KPIs relevantes**

- **Definición:** Los KPIs son métricas específicas que ayudan a medir el progreso hacia los objetivos de una organización.
- **Ejemplo:** En una empresa de logística, un KPI puede ser "reducción del tiempo promedio de entrega en un 20%".
- **Cómo diseñar un buen KPI:**
  - Específico: ¿Qué se está midiendo?
  - Medible: ¿Cómo se mide?
  - Alcanzable: ¿Es posible lograrlo?
  - Relevante: ¿Está alineado con los objetivos organizacionales?
  - Basado en tiempo: ¿Cuál es el plazo para alcanzarlo?

Con estos recursos y conceptos, podrás analizar y proponer soluciones efectivas para el caso del SEAD UAM.

### **3. Herramientas, estrategias y caso de estudio (60 minutos)**

#### **Caso de estudio: SEAD UAM**

**Descripción vivencial:**
Imagina que estás trabajando en el Departamento de Estudios a Distancia de la Universidad Autónoma de Manizales (SEAD UAM) en el año 2020. El equipo enfrenta problemas recurrentes debido a la fragmentación de la información y la falta de un enfoque analítico consolidado. En tu día a día, observas cómo:

1. **Sistemas de información fragmentados:**

   - La información se encuentra dispersa en múltiples hojas de cálculo y sistemas no conectados. Cada departamento maneja datos de manera aislada, generando inconsistencias y retrasos.
   - Ejemplo: Dos áreas reportan cifras diferentes sobre la misma métrica de inscripciones, lo que causa confusión en las decisiones de marketing.

2. **Procesos manuales y propensos a errores:**

   - La recopilación de datos para los informes mensuales implica largas jornadas de trabajo, copiando y pegando información desde distintos sistemas.
   - Ejemplo: Al integrar los datos manualmente, el equipo detecta duplicados y errores en las cifras de desempeño académico.

3. **Ausencia de una cultura de datos:**

   - Las decisiones suelen tomarse por intuición, sin respaldo en métricas claras ni análisis detallado. El equipo percibe una resistencia al cambio en los procesos tradicionales.
   - Ejemplo: La falta de indicadores clave impide evaluar el impacto de las campañas educativas implementadas.

**Situación inicial para los estudiantes:**
Eres parte del equipo de datos del SEAD UAM en este contexto. Tu responsabilidad es identificar las áreas clave que requieren mejoras y diseñar un plan estratégico para aumentar la madurez analítica del departamento. El futuro del SEAD UAM depende de tu capacidad para proponer soluciones efectivas.

**Actividad práctica:**

1. **Ejercicio grupal:**

   - Analiza la situación inicial del SEAD UAM considerando los problemas enumerados.
   - Diseña soluciones en las siguientes áreas:
     - **Centralización de Datos:** ¿Cómo puedes integrar todos los datos en una única plataforma para evitar inconsistencias?
     - **Automatización de Procesos:** ¿Qué herramientas implementarías para reducir errores y optimizar el tiempo de análisis?
     - **Cultura de Datos:** ¿Qué estrategias propondrías para fomentar una toma de decisiones basada en evidencia?

2. **Discusión:**

   - Compara tus propuestas con las acciones reales implementadas por el SEAD UAM:
     - Desarrollo de un sistema integrado de gestión de información que centraliza datos de estudiantes, cursos y desempeño.
     - Automatización de informes clave mediante el uso de herramientas de inteligencia de negocios.
     - Implementación de programas de capacitación para el equipo, promoviendo el uso de datos en la planificación estratégica.

**Resultados obtenidos por el SEAD UAM tras la mejora:**

- Mejora en la calidad y consistencia de la información.
- Reducción de tiempos en la obtención y análisis de datos.
- Toma de decisiones más informada y basada en datos.
