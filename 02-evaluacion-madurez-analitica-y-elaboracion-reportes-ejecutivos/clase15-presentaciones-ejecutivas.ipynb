## 3. Storytelling con evidencia: una secuencia para decidir

![Secuencia desde observación hasta control de la recomendación](assets/03_cadena_de_argumento.png)

Un argumento sólido distingue cinco niveles:

| Nivel | Responde | Cuidado con… |
|---|---|---|
| **Observación** | ¿Qué ocurrió? ¿Cuándo? ¿Cuánto? | Confundir una estimación con un hecho. |
| **Interpretación** | ¿Qué implica? ¿Qué explicaciones compiten? | Convertir correlación en causalidad. |
| **Alternativas** | ¿Qué opciones reales hay? | Presentar solo la favorita. |
| **Solicitud** | ¿Qué decisión, por cuánto y hasta cuándo? | Una recomendación vaga o no reversible. |
| **Control** | ¿Cómo sabremos si funcionó? | Omitir supuestos, dueño o regla de parada. |

**Transformación de titulares (3 minutos):**

- Descriptivo: «Costos de atención».
- Exagerado: «La sucursal Norte está destruyendo el negocio».
- Verificable: «En septiembre, Norte tardó 18 minutos frente a una meta de 12».

Redacta un cuarto título que incluya una **solicitud concreta** sin afirmar que ya se demostró una causa.

**La recomendación va temprano cuando hablamos ante un comité ejecutivo.** En esta clase, la solución del caso aparecerá *después* de revisar la evidencia porque necesitamos aprender a construirla.

## 4. Visualización profesional: claridad sin distorsión

![Barras con eje recortado frente a barras desde cero](assets/04_graficas_antes_despues.png)

**Diagnóstico en parejas (2 minutos):** ¿qué percepción exagera la gráfica de la izquierda? ¿La de la derecha sería suficiente para decidir?

**Prácticas fundamentales:**

- Usa **posición y longitud** cuando el objetivo sea comparar valores. Las barras deben tener **base cero**; una serie temporal de línea puede usar una escala enfocada si está explícita y no induce a error.
- Muestra **unidad, periodo, población, denominador, referencia** y fuente. Un porcentaje sin base puede engañar.
- Usa color, anotaciones y ordenación para **dirigir la atención a evidencia**, no para manipularla.
- Identifica visualmente **observado**, **modelado** y **supuesto**. No llames «confianza del 95 %» a tres escenarios inventados.
- Deja detalle técnico para un anexo cuando sea necesario, sin eliminar límites ni resultados desfavorables.

No existe una regla científica universal de «tres colores» o «seis bullets». La cantidad adecuada depende de la pregunta y de la complejidad de la información.

## 5. Caso NovaCrédito: una decisión con información imperfecta

![Ficha narrativa del caso NovaCrédito sin revelar la solución](assets/07_ficha_novacredito.png)

**Situación:** NovaCrédito, fintech ficticia, adquiere clientes de crédito de consumo mediante **Social, Búsqueda y Referidos**. En octubre de 2026, el comité debe decidir el presupuesto del siguiente trimestre. Existe una propuesta preliminar de **considerar hasta $90,000 MXN** para una posible reasignación, todavía **sin aprobar**.

**Tres opciones defendibles:**

- **A. Mantener** la mezcla temporalmente, hacer un diagnóstico y fijar fecha de decisión.
- **B. Piloto reversible** por un monto y plazo acotados, sujeto a controles.
- **C. Reasignación amplia** con justificación adicional.

**Dilema:** ¿el CAC promedio más bajo indica también la mejor decisión marginal? ¿Qué pasa si ese canal no puede crecer, o si su calidad crediticia fuera peor?

**Recibirás cuatro fuentes de datos:** adquisición mensual 2026, calidad histórica de cohortes maduras 2025, capacidad Q4 preliminar y escenarios de costos marginales hipotéticos. **No** todas hablan del mismo periodo ni tienen la misma certeza.

```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

DATA = Path('data')
assert DATA.is_dir(), 'Abre el notebook desde la carpeta raíz del paquete (junto a datos/).'

adq = pd.read_csv(DATA / 'adquisicion_mensual.csv', parse_dates=['mes'])
calidad = pd.read_csv(DATA / 'calidad_cohortes_maduras.csv')
capacidad = pd.read_csv(DATA / 'capacidad_operativa.csv')
supuestos = pd.read_csv(DATA / 'supuestos_escenarios.csv')

assert len(adq) == 18 and not adq.duplicated(['mes', 'canal']).any()
assert (adq['solicitudes'] >= adq['altas']).all()
assert (calidad['clientes_maduros_n'] >= calidad['clientes_90dpd_n']).all()
print(f'Datos preparados | Python {sys.version.split()[0]} | pandas {pd.__version__}')
print(f'{len(adq)} observaciones de adquisición · {len(calidad)} cohortes de calidad')
display(adq.head(6))
```

### Primera lectura: 3 preguntas, no 30 KPIs

1. ¿Cuál canal cambió más su **costo por alta** entre abril y septiembre?
2. ¿Qué canal tiene mejor desempeño en el **último trimestre**? Usa CAC ponderado: `sum(gasto) / sum(altas)`.
3. ¿Es defendible cambiar presupuesto usando solo estos datos? ¿Qué información falta?

**Horizon / denominadores:** el CAC es **gasto del canal / altas atribuidas**, no costo incremental causal. El margen a 12 meses que encontrarás abajo es una **proyección de modelo por alta**, no margen realizado.

```python
adq = adq.assign(
    cac_mxn=lambda x: x['gasto_mxn'] / x['altas'],
    tasa_conversion=lambda x: x['altas'] / x['solicitudes'],
    margen_12m_total_estimado_mxn=lambda x: x['altas'] * x['margen_12m_estimado_por_alta_mxn'],
)
q3 = adq.loc[adq['mes'].between('2026-07-01', '2026-09-01')]
resumen = q3.groupby('canal', as_index=False).agg(
    gasto_mxn=('gasto_mxn', 'sum'),
    altas=('altas', 'sum'),
    solicitudes=('solicitudes', 'sum'),
    margen_12m_total_estimado_mxn=('margen_12m_total_estimado_mxn', 'sum'),
)
resumen['cac_ponderado_mxn'] = resumen['gasto_mxn'] / resumen['altas']
resumen['conversion'] = resumen['altas'] / resumen['solicitudes']
resumen['contribucion_neta_12m_estimada_mxn'] = (
    resumen['margen_12m_total_estimado_mxn'] - resumen['gasto_mxn']
)
resumen['contribucion_estimada_por_alta_mxn'] = (
    resumen['contribucion_neta_12m_estimada_mxn'] / resumen['altas']
)
display(resumen.round({'cac_ponderado_mxn': 0, 'conversion': 3,
                       'contribucion_estimada_por_alta_mxn': 0}))
assert resumen['altas'].sum() == q3['altas'].sum()
```

```python
# Mismo dataset; una visualización diseñada para la pregunta: ¿qué cambió?
fig, ax = plt.subplots(figsize=(10, 4.5))
colores = {'Búsqueda':'#356D9A', 'Social':'#BB514D', 'Referidos':'#297F7C'}
for canal, g in adq.groupby('canal'):
    g = g.sort_values('mes')
    ax.plot(g['mes'], g['cac_mxn'], label=canal, color=colores[canal],
            marker='o', lw=2.4)
    ax.annotate(f"${g.iloc[-1]['cac_mxn']:,.0f}",
                xy=(g.iloc[-1]['mes'], g.iloc[-1]['cac_mxn']),
                xytext=(7, 0), textcoords='offset points', va='center', fontsize=10)
ax.set(title='Evolución del CAC observado por canal · abril–septiembre 2026',
       ylabel='MXN por alta atribuida', xlabel='Mes')
ax.grid(alpha=.18)
ax.legend(loc='upper left', frameon=False)
ax.margins(x=.11)
plt.tight_layout()
plt.show()
```

### Una fuente que contradice la lectura simple

El equipo de Riesgo aporta **cohortes maduras de 2025** con su número observado de clientes que alcanzaron 90+ días de mora. **No** corresponden a las adquisiciones recientes de 2026. La comparación es una **señal de vigilancia**, no una comparación causal ajustada por mezcla de clientes.

**Discute:** si Referidos tiene bajo CAC, pero una mora histórica mayor, ¿cómo cambia el argumento ante el CFO? ¿Qué habría que validar?

```python
riesgo = calidad.groupby('canal', as_index=False).agg(
    maduros_n=('clientes_maduros_n', 'sum'),
    mora_90d_n=('clientes_90dpd_n', 'sum'),
)
riesgo['mora_90d_observada'] = riesgo['mora_90d_n'] / riesgo['maduros_n']
fig, ax = plt.subplots(figsize=(9, 3.5))
r = riesgo.sort_values('mora_90d_observada')
bars = ax.barh(r['canal'], 100*r['mora_90d_observada'],
               color=[colores[c] for c in r['canal']])
ax.bar_label(bars, labels=[f"{v:.1%} (n={n})" for v,n in
                            zip(r['mora_90d_observada'], r['maduros_n'])], padding=5)
ax.set_xlim(0, max(100*r['mora_90d_observada'])*1.65)
ax.set(xlabel='% de clientes maduros con 90+ DPD',
       title='Mora observada en cohortes maduras 2025 (distinta población)')
ax.grid(axis='x',alpha=.15)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
display(capacidad[['canal','altas_incrementales_max_q4','nivel_validacion']])
```

### Supuestos para comparar un posible piloto (no pronóstico)

Una opción es **considerar** mover *hasta* $90,000 en Q4 de Social a Referidos. Pero los datos revelan que la capacidad operativa de Referidos está **sin validar**. Aunque el comité autorice un máximo, quizá solo pueda reasignarse parte del dinero.

En la celda siguiente comparamos tres **escenarios subjetivos** con costo marginal, margen futuro **proyectado**, capacidad y costo extra de implementación. No son intervalos de confianza ni resultados de un A/B test.

**Lógica económica:** para cada peso realmente movido comparamos el margen futuro modelado de nuevas altas en Referidos con el margen modelado de altas que dejarían de adquirirse en Social; el presupuesto de marketing transferido es igual en ambas opciones y se cancela. Restamos aparte el costo incremental de implementar el piloto. Todo depende de hipótesis aún sin validar.

```python
presupuesto_maximo_mxn = 90_000  # máximo propuesto, no monto comprometido
esc = supuestos.copy()
esc['monto_trasladado_factible_sup_mxn'] = np.minimum(
    presupuesto_maximo_mxn,
    esc['cupo_ref_adicional_sup_n'] * esc['cac_marginal_ref_sup_mxn'],
)
esc['altas_ref_incrementales_hipoteticas'] = (
    esc['monto_trasladado_factible_sup_mxn'] / esc['cac_marginal_ref_sup_mxn']
)
esc['altas_social_desplazadas_hipoteticas'] = (
    esc['monto_trasladado_factible_sup_mxn'] / esc['cac_marginal_social_sup_mxn']
)
esc['delta_contribucion_12m_estimada_mxn'] = (
    esc['altas_ref_incrementales_hipoteticas'] * esc['margen_12m_ref_sup_mxn']
    - esc['altas_social_desplazadas_hipoteticas'] * esc['margen_12m_social_sup_mxn']
    - esc['costo_implementacion_sup_mxn']
)
esc['presupuesto_sin_reasignar_mxn'] = presupuesto_maximo_mxn - esc['monto_trasladado_factible_sup_mxn']
assert (esc['presupuesto_sin_reasignar_mxn'] >= 0).all()
assert (esc['altas_ref_incrementales_hipoteticas'] <= esc['cupo_ref_adicional_sup_n']).all()
display(esc[['escenario','monto_trasladado_factible_sup_mxn',
             'presupuesto_sin_reasignar_mxn','altas_ref_incrementales_hipoteticas',
             'delta_contribucion_12m_estimada_mxn']].round(0))

fig, ax = plt.subplots(figsize=(9,3.5))
vals = esc['delta_contribucion_12m_estimada_mxn'] / 1000
bars = ax.barh(esc['escenario'], vals,
               color=['#BB514D' if v < 0 else '#297F7C' for v in vals])
ax.axvline(0,color='#193248',lw=1)
for rect, value in zip(bars,vals):
    ax.annotate(f'{value:+.1f} mil', xy=(value,rect.get_y()+rect.get_height()/2),
                xytext=(5 if value>=0 else -5,0),textcoords='offset points',
                va='center',ha='left' if value>=0 else 'right')
ax.set(xlabel='Cambio hipotético en contribución a 12m, miles de MXN',
       title='Sensibilidad del piloto a supuestos y límites de capacidad')
ax.margins(x=.25)
ax.grid(axis='x',alpha=.15)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

### Decisión bajo incertidumbre: tú propones

Antes de crear las láminas, completa oralmente estas cuatro frases:

1. **Hecho:** «En [periodo], [métrica] fue [cifra] frente a [comparador]».
2. **Contrapunto:** «Sin embargo, [riesgo o dato que complica la decisión]».
3. **Hipótesis:** «Nuestra propuesta podría funcionar *si* [supuesto que debe verificarse]».
4. **Solicitud:** «Pedimos a [audiencia] autorizar / posponer / investigar [acción, límite y plazo]».

**No copies una recomendación prefabricada:** A, B y C pueden defenderse si estableces un criterio claro, su costo de oportunidad, su riesgo y qué nueva evidencia haría cambiar de decisión.
