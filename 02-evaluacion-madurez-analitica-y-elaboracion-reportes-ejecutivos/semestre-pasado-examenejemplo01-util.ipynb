```python
import numpy as np
import pandas as pd

# Set a random seed for reproducibility
np.random.seed(42)

# Define the number of months (2 years)
months = pd.date_range(start="2022-01-01", periods=24, freq="M")

# Marketing Expenditure (USD) - trend where marketing increases over time
marketing_expenditure = np.random.normal(loc=20000, scale=3000, size=len(months)) + np.linspace(0, 5000, len(months))

# Web Traffic - higher marketing leads to more web traffic with some randomness
web_traffic = np.round(marketing_expenditure / 2 + np.random.normal(loc=10000, scale=2000, size=len(months)),0)

# Customer Satisfaction (CSAT) - mostly stable, but with small random fluctuations
customer_satisfaction = np.random.normal(loc=4.2, scale=0.2, size=len(months))

# Churn Rate (%) - tends to increase if customer satisfaction decreases
churn_rate = 0.02 + (4.5 - customer_satisfaction) * 0.01 + np.random.normal(loc=0, scale=0.005, size=len(months))

# Product Price (USD) - slight monthly variation in product pricing
product_price = np.random.normal(loc=100, scale=5, size=len(months))

# Advertising Effectiveness (%) - related to how well the marketing expenditure converts into web traffic and new customers
advertising_effectiveness = np.random.normal(loc=0.7, scale=0.05, size=len(months))

# Customer Service Interactions - related to the number of customers, higher interaction for higher customer base
customer_service_interactions = np.random.normal(loc=500, scale=100, size=len(months)) + np.linspace(0, 300, len(months))

# Returning Customers (%) - higher customer satisfaction and lower churn rate result in more returning customers
returning_customers_rate = 0.2 + customer_satisfaction * 0.05 - churn_rate * 0.1 + np.random.normal(loc=0, scale=0.02, size=len(months))

# Economic Conditions (index 0.9 to 1.1) - positive/negative economic environment can affect revenue
economic_conditions = np.random.normal(loc=1, scale=0.05, size=len(months))

# Seasonality Factor - higher sales in certain months (e.g., December)
seasonality = 1 + 0.2 * np.sin(2 * np.pi * (months.month - 1) / 12)

# Promotions Effectiveness - occasional promotions can increase sales
promotions = np.random.choice([0, 0.1, 0.15, 0.2], size=len(months), p=[0.6, 0.2, 0.15, 0.05])

# New Customers - related to web traffic, churn rate, and advertising effectiveness
new_customers = (web_traffic * (1 - churn_rate) * advertising_effectiveness) / 100 + np.random.normal(loc=50, scale=20, size=len(months))

# Existing customers (calculated cumulatively, affected by churn and new customers)
existing_customers = np.zeros(len(months))
existing_customers[0] = 1000  # Starting number of customers
for i in range(1, len(existing_customers)):
    existing_customers[i] = existing_customers[i-1] + new_customers[i] - churn_rate[i] * existing_customers[i-1]

# Returning Customers - proportion of existing customers that are returning based on the returning customer rate
returning_customers = existing_customers * returning_customers_rate

# Total Revenue - related to new and returning customers, product price, customer satisfaction, economic conditions, seasonality, and promotions
total_revenue = (
    (new_customers + returning_customers) * product_price * economic_conditions * seasonality * (1 + promotions)
    + customer_satisfaction * 1000
    + np.random.normal(loc=0, scale=5000, size=len(months))
)

# Combine the variables into a DataFrame
data = pd.DataFrame({
    'Date': months,
    'Marketing_Expenditure': marketing_expenditure,
    'Web_Traffic': web_traffic,
    'Customer_Satisfaction': np.round(customer_satisfaction,2),
    'Churn_Rate': churn_rate,
    'Product_Price': np.round(product_price,2),
    'Customer_Service_Interactions': np.round(customer_service_interactions,2),
    'Economic_Conditions': economic_conditions,
    'Promotions': promotions,
    'New_Customers': np.round(new_customers,0),
    'Existing_Customers': np.round(existing_customers,0),
    'Returning_Customers': np.round(returning_customers,0),
    'Total_Revenue': total_revenue
})

# Show the first few rows of the dataset
print(data.head())

# Save the synthetic dataset to a CSV file
data.to_csv('data/optitech_solutions_sales.csv', index=False)
```

```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Define the number of customers to generate
num_customers = 1518

# Generate Customer ID
customer_id = np.arange(1, num_customers + 1)

# Generate Edad del Niño (between 4 and 17 years old)
edad_del_nino = np.random.randint(4, 18, size=num_customers)

# Generate Sexo del Niño (0 = Female, 1 = Male)
sexo_del_nino = np.random.choice([0, 1], size=num_customers, p=[0.45, 0.55])

# Generate Nivel de Estudios del Padre (Categorical: 0 = High School, 1 = Undergraduate, 2 = Graduate)
nivel_estudios_padre = np.random.choice([0, 1, 2], size=num_customers, p=[0.3, 0.4, 0.3])

# Generate Ingresos del Hogar (annual household income in USD)
ingresos_hogar = np.round(np.random.normal(50000, 15000, size=num_customers), 2)
ingresos_hogar = np.clip(ingresos_hogar, 20000, 150000)  # Household income capped between 20k and 150k

# Generate Plan de Subscripción (more expensive plans chosen by wealthier households)
plan_subscripcion = np.zeros(num_customers, dtype=object)
for i in range(num_customers):
    if ingresos_hogar[i] < 40000:
        plan_subscripcion[i] = np.random.choice(['Básico'], p=[1.0])
    elif 40000 <= ingresos_hogar[i] <= 80000:
        plan_subscripcion[i] = np.random.choice(['Básico', 'Premium'], p=[0.6, 0.4])
    else:
        plan_subscripcion[i] = np.random.choice(['Básico', 'Premium', 'Familiar'], p=[0.3, 0.3, 0.4])

# Generate Duración de la Subscripción (correlated with churn; churners tend to have shorter subscriptions)
duracion_subscripcion = np.zeros(num_customers)
for i in range(num_customers):
    if plan_subscripcion[i] == 'Básico':
        duracion_subscripcion[i] = np.random.randint(1, 25)
    elif plan_subscripcion[i] == 'Premium':
        duracion_subscripcion[i] = np.random.randint(6, 36)
    else:  # Familiar plan
        duracion_subscripcion[i] = np.random.randint(12, 36)

# Generate Número de Sesiones por Semana (correlated with Plan, Edad del Niño, and Sexo del Niño)
sesiones_por_semana = np.zeros(num_customers)
for i in range(num_customers):
    if plan_subscripcion[i] == 'Básico':
        sesiones_por_semana[i] = np.random.normal(2 + edad_del_nino[i] / 15, 1)
    elif plan_subscripcion[i] == 'Premium':
        sesiones_por_semana[i] = np.random.normal(4 + edad_del_nino[i] / 10 + sexo_del_nino[i], 1.5)
    else:  # Familiar plan
        sesiones_por_semana[i] = np.random.normal(6 + edad_del_nino[i] / 7 + sexo_del_nino[i], 2)
sesiones_por_semana = np.clip(sesiones_por_semana, 1, 10)

# Generate Nivel de Satisfacción del Cliente (correlated with Plan and Sesiones por Semana)
nivel_satisfaccion = np.zeros(num_customers)
for i in range(num_customers):
    if plan_subscripcion[i] == 'Básico':
        nivel_satisfaccion[i] = np.random.normal(3 + sesiones_por_semana[i] / 5, 0.5)
    elif plan_subscripcion[i] == 'Premium':
        nivel_satisfaccion[i] = np.random.normal(4 + sesiones_por_semana[i] / 8, 0.4)
    else:  # Familiar plan
        nivel_satisfaccion[i] = np.random.normal(4.5 + sesiones_por_semana[i] / 10, 0.3)
nivel_satisfaccion = np.clip(nivel_satisfaccion, 1, 5)

# Generate Churn (based on satisfaction, income, and subscription duration, 1 = churned, 0 = active)
churn = np.zeros(num_customers)
churn_prob = 0.1 + (4 - nivel_satisfaccion) * 0.15 + (1 - duracion_subscripcion / 36) + np.random.normal(0, 0.05, size=num_customers)
churn = np.where(churn_prob > 0.5, 1, 0)

# Generate Gasto Total en Subscripción (dependent on plan, duration, and churn)
gasto_total_subscripcion = np.zeros(num_customers)
for i in range(num_customers):
    if plan_subscripcion[i] == 'Básico':
        costo_mensual = 10
    elif plan_subscripcion[i] == 'Premium':
        costo_mensual = 20
    else:
        costo_mensual = 30
    gasto_total_subscripcion[i] = costo_mensual * duracion_subscripcion[i] * (1 if churn[i] == 0 else 0.7)

# Generate Valor del Tiempo de Vida (LTV, dependent on satisfaction, churn, household income, studies, and sex of the kid)
ltv = np.zeros(num_customers)
for i in range(num_customers):
    estudio_factor = 1 + nivel_estudios_padre[i] * 0.1  # Higher level of studies leads to a slightly higher LTV
    sexo_factor = 1.1 if sexo_del_nino[i] == 1 else 0.9  # Boys tend to play more, increasing LTV
    ltv[i] = (gasto_total_subscripcion[i] + nivel_satisfaccion[i] * 50 + ingresos_hogar[i] * 0.01) * estudio_factor * sexo_factor

# Create DataFrame
data = pd.DataFrame({
    'ID del Cliente': customer_id,
    'Edad del Niño': edad_del_nino,
    'Sexo del Niño': np.where(sexo_del_nino == 1, 'Masculino', 'Femenino'),
    'Nivel de Estudios del Padre': np.where(nivel_estudios_padre == 0, 'High School', 
                                           np.where(nivel_estudios_padre == 1, 'Undergraduate', 'Graduate')),
    'Duración de la Subscripción (meses)': duracion_subscripcion,
    'Plan de Subscripción': plan_subscripcion,
    'Ingresos del Hogar (USD)': ingresos_hogar,
    'Gasto Total en Subscripción (USD)': np.round(gasto_total_subscripcion, 2),
    'Número de Sesiones por Semana': np.round(sesiones_por_semana, 2),
    'Nivel de Satisfacción del Cliente': np.round(nivel_satisfaccion, 2),
    'Churn': churn,
    'Valor del Tiempo de Vida (LTV)': np.round(ltv, 2)
})

# Display the first few rows of the dataset
print(data.head())

# Save the synthetic dataset to a CSV file
data.to_csv('data/gabu_data.csv', index=False)

print("Synthetic dataset created and saved as 'gabu_synthetic_data_v3.csv'")
```
