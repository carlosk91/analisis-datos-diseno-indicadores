### **2. Introducción a Estadísticos Descriptivos (30 minutos):**

- **Definición y conceptos clave:**
  - Medidas de tendencia central: media, mediana, moda.
  - Medidas de dispersión: rango, varianza, desviación estándar.
  - Importancia de cada medida en el análisis de datos.

- **Ejemplo:**
  - Cómo calcular la media y desviación estándar de las ventas usando Python:

```python
import pandas as pd

data = pd.read_csv("data/modaexpress_ventas.csv")
```

```python
data.info()
```

```python
data.head()
```

```python
print(f"Media: {data['Ventas (USD)'].mean()}, Desviación Estándar: {data['Ventas (USD)'].std()}")
```

```python
data.describe()
```

```python
data['Gasto Proporcional en Promociones'] = data['Gasto en Promociones (USD)'] / data['Ventas (USD)']
```

- **Actividad guiada (15 minutos):**
  - Calcular medidas centrales y de dispersión para las ventas, clientes y promociones en cada región. 
  - Discusión: ¿Cuál es la región con mayor variabilidad en ventas? ¿Qué podría significar esto para la empresa?

```python
df_gastos = data.groupby("Región").agg({'Ventas (USD)': ['mean', 'std'], \
                                        'Gasto en Promociones (USD)': ['mean', 'std'], \
                                        'Clientes Atendidos': ['mean', 'std'], \
                                        'Gasto Proporcional en Promociones': ['mean', 'std']})
```

```python
df_gastos
```

- ¿Qué región tiene mayor gasto proporcional en Promociones?

```python
df_gastos.sort_values(by=[('Gasto Proporcional en Promociones','mean')], ascending=False)
```

- ¿Qué región tiene mayor variabilidad en Ventas?

```python
df_gastos.sort_values(by=[('Ventas (USD)','std')], ascending=False)
```

### **3. Distribuciones de Datos (30 minutos):**

- Analicemos las ventas

- **Conceptos clave:**
  - Tipos de distribuciones: normal, uniforme, sesgada a la derecha/izquierda.
  - Visualización de distribuciones:
    - Histogramas.
    - Gráficas KDE (Kernel Density Estimation).

- **Ejemplo:**
  - Generar un histograma para visualizar las ventas:

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Histograma de ventas
sns.histplot(data=data, x="Ventas (USD)", kde=True)
plt.title("Distribución de Ventas")
plt.xlabel("Ventas (USD)")
plt.ylabel("Frecuencia")
plt.show()
```

### ¿Hay diferencias en proporción de productos vendidos por región?

```python
df_producto = data.groupby(["Región","Categoría de Producto"]).agg({'Ventas (USD)': ['mean', 'sum'], \
                                        'Gasto en Promociones (USD)': ['mean', 'std'], \
                                        'Clientes Atendidos': ['mean', 'std'], \
                                        'Gasto Proporcional en Promociones': ['mean', 'std']})
```

```python
ventas_por_producto = data.groupby("Región")["Ventas (USD)"].sum()
```

```python
for region in ventas_por_producto.index:
    region_total = ventas_por_producto[region]
    for category in df_producto.loc[region].index:
        proportion = df_producto.loc[(region, category), ('Ventas (USD)', 'sum')] / region_total

        df_producto.loc[(region, category), ('Proporción de Ventas en Región', '')] = proportion.round(3)
```

```python
df_producto.head()
```

```python
plt.figure(figsize=(8, 4))

sns.barplot(x='Región', y='Proporción de Ventas en Región', hue='Categoría de Producto', data=df_producto, palette='viridis')

plt.title('Proporción de Ventas por Categoría de Producto y Región', fontsize=14)
plt.xlabel('Región')
plt.ylabel('Proporción de Ventas')
plt.xticks(rotation=0)
plt.ylim(0, 0.4)
plt.legend(title='Categoría de Producto', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
