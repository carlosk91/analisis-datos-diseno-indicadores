## Reto 4 — Entender el desempeño y calidad de soporte de los conductores (25 puntos)

### Contexto

Panda Cosmos quiere entender el desempeño y la calidad de soporte de sus operadores, así como métricas adicionales.

### Dataset: `panda_cosmos_operadores.csv`

### Columnas:

* conductor_id: Identificador único asignado a cada conductor en la plataforma.
* tiempo_promedio_entrega: Tiempo promedio en minutos que el conductor tarda en completar una entrega.
* entregas_completadas: Número total de entregas exitosas realizadas por el conductor en el último trimestre.
* valoracion_cliente: Calificación promedio (de 1 a 5) otorgada por los clientes al conductor.
* latencia_respuesta_app: Tiempo promedio, en segundos, que tarda el conductor en responder una notificación de la app.
* tasa_incidencias: Proporción de entregas que registraron algún tipo de incidente o reporte por parte del cliente.
* retencion_anual: Indicador binario que señala si el conductor continúa activo después de un año (1 = sí, 0 = no).

### Parte A: Modelo SEM (10 puntos)

* Define dos variables latentes:

  * *Desempeño Operativo* → `tiempo_promedio_entrega`, `entregas_completadas`, `valoracion_cliente`
  * *Calidad de Soporte* → `latencia_respuesta_app`, `tasa_incidencias`
* Modela con `semopy`, visualiza y reporta coeficientes y cargas factoriales.

### Parte B: Clustering (10 puntos)

* Usa `KMeans` con las variables operativas (`tiempo_promedio_entrega`, `entregas_completadas`, `valoracion_cliente`) para agrupar a los conductores.
* Identifica el número óptimo de clústeres usando el método del codo.
* Nombra cada grupo según sus características y grafica la distribución.

### Parte C: Acciones (5 puntos)

* Compara la retención anual entre clústeres.
* Sugiere dos intervenciones para mejorar el desempeño operativo y la calidad de soporte en los grupos con bajo desempeño.
