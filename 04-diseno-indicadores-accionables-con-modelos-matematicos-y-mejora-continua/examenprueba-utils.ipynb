```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

# 1. Generate Sales Data for Time Series Forecast
np.random.seed(42)

# Create a date range for 3 years of monthly data
date_range = pd.date_range(start="2020-01-01", end="2022-12-01", freq="MS")
# Generate sales data with a seasonal pattern and some noise
seasonal_pattern = [1000 + 200 * np.sin(2 * np.pi * (month / 12)) for month in range(36)]
sales = np.array(seasonal_pattern) + np.random.normal(0, 100, len(seasonal_pattern))
sales_data = pd.DataFrame({"Fecha": date_range, "Ventas": sales})
sales_data.to_csv("data/ventas_mensuales.csv", index=False)


# 2. Generate Customer Satisfaction Data for SEM
num_customers = 300

# Simulate variables for customer satisfaction model
np.random.seed(42)
uso = np.random.normal(4, 0.5, num_customers)
seguridad = uso + np.random.normal(0, 0.3, num_customers)
recomendaciones = uso * 0.8 + np.random.normal(0, 0.2, num_customers)
tiempo_respuesta = np.random.normal(4.2, 0.3, num_customers)
resolucion_problemas = tiempo_respuesta * 0.9 + np.random.normal(0, 0.2, num_customers)
tamaño_empresa = np.random.randint(1, 4, num_customers)  # Small, Medium, Large coded as 1, 2, 3
satisfaccion = 0.5 * uso + 0.7 * tiempo_respuesta + np.random.normal(0, 0.5, num_customers)

# Create a DataFrame
satisfaction_data = pd.DataFrame({
    "Uso": uso,
    "Seguridad": seguridad,
    "Recomendaciones": recomendaciones,
    "Tiempo_Respuesta": tiempo_respuesta,
    "Resolucion_Problemas": resolucion_problemas,
    "Tamaño_Empresa": tamaño_empresa,
    "Satisfaccion": satisfaccion
})
satisfaction_data.to_csv("data/satisfaccion_cliente.csv", index=False)

import pandas as pd
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Define parameters for each group
n_users_A = 10000  # Total users in Group A
n_users_B = 10000  # Total users in Group B
conversion_rate_A = 0.06  # 6% conversion rate for Group A
conversion_rate_B = 0.055  # 5.5% conversion rate for Group B
price_A = 50  # Price for Group A
price_B = 60  # Price for Group B

# Simulate conversions (1 for converted, 0 for not converted)
conversions_A = np.random.choice([1, 0], size=n_users_A, p=[conversion_rate_A, 1 - conversion_rate_A])
conversions_B = np.random.choice([1, 0], size=n_users_B, p=[conversion_rate_B, 1 - conversion_rate_B])

# Calculate revenue per user
revenue_A = conversions_A * price_A
revenue_B = conversions_B * price_B

# Create datasets
data_A = pd.DataFrame({
    'Group': 'A',
    'Converted': conversions_A,
    'Revenue': revenue_A
})

data_B = pd.DataFrame({
    'Group': 'B',
    'Converted': conversions_B,
    'Revenue': revenue_B
})

# Combine datasets
ab_test_data = pd.concat([data_A, data_B]).reset_index(drop=True)
ab_test_data.to_csv("data/ab_test_data.csv", index=False)
```
