```python
import pandas as pd 
import numpy as np
```

```python
people_control = 500
people_treatment = 500

conversions_control = [1 if np.random.random() < 0.15 else 0 for _ in range(people_control)]
conversions_treatment = [1 if np.random.random() < 0.09 else 0 for _ in range(people_treatment)]

rate_control = len(conversions_control) / people_control
rate_treatment = len(conversions_treatment) / people_treatment

ltv_control = np.random.normal(100, 10, people_control)
ltv_treatment = np.random.normal(130, 10, people_treatment)

cost_of_acquisition = np.random.normal(50, 3, people_control)
cost_of_acquisition_treatment = np.random.normal(50, 3, people_treatment)

df_control = pd.DataFrame({
    "conversion": conversions_control,
    "ltv": ltv_control,
    "cost_of_acquisition": cost_of_acquisition
})

df_treatment = pd.DataFrame({
    "conversion": conversions_treatment,
    "ltv": ltv_treatment,
    "cost_of_acquisition": cost_of_acquisition_treatment
})
```
